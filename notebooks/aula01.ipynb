## Contas: as quatro que você vai repetir

````python
# Divisao, divisao inteira, resto e potenciacao.
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 10)

# Um arredondamento explicito evita a cauda de casas decimais.
print(10 / 3)
print(round(10 / 3, 2))

# Segundos fora do ar -> HH:MM:SS
total = 3661
print(total // 3600, total % 3600 // 60, total % 60)
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

| Operador | O que faz | Detalhe importante |
|---|---|---|
| `/` | divisão | **sempre** devolve `float`, mesmo em `4 / 2` |
| `//` | divisão inteira | descarta a parte fracionária |
| `%` | resto | a dupla `//` e `%` resolve hora, minuto e segundo |
| `**` | potenciação | `10 ** 3`, não `pow` nem `^` |

Repare que `/` preserva a parte fracionária mesmo entre dois inteiros: `7 / 2` dá
`3.5`, e não `3`. Quando você quer o `3` — quantas fibras inteiras cabem, quantos
racks completos, quantas horas fechadas —, peça explicitamente com `//`. Usar `/`
onde se queria `//` é o tipo de erro que não aparece: o programa roda e entrega
`3.5` slots.

`round(valor, casas)` aparece em quase todo exercício deste curso, e por um
motivo prático: `10 / 3` tem dezesseis casas decimais que ninguém quer ver num
relatório de potência.

`//` tem um detalhe que só aparece com número negativo: ele não corta a parte
decimal, **arredonda para baixo**.

</details>

````python
# // nao corta a parte decimal: ele arredonda PARA BAIXO.
print(5 // 2)
print(-5 // 2)
print(5 % 2)
print(-5 % 2)
````

## A conta que abre qualquer projeto de telecom

Potência óptica não se mede em miliwatts: mede-se em **dBm**, a mesma grandeza
numa escala logarítmica. A razão é conveniência — em dB, ganhos e perdas ao longo
do enlace viram soma e subtração em vez de multiplicação e divisão.

````python
import math

# mW -> dBm e o caminho de volta.
potencia_mw = 0.001
potencia_dbm = 10 * math.log10(potencia_mw)
print(potencia_dbm)

de_volta = 10 ** (potencia_dbm / 10)
print(de_volta)

# 3 dB a mais e o dobro da potencia -- a conta que todo mundo faz de cabeca.
print(round(10 * math.log10(2), 2))
print(round(10 * math.log10(4), 2))
print(round(10 * math.log10(8), 2))
````

**Preveja:** "o dobro da potência dá quantos dB a mais? E quatro vezes?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Guarde o último número: **3 dB é o dobro da potência**. Um enlace que perde 3 dB
entregou metade do que recebeu. E os 3 dB de cada dobra vão se somando: quatro
vezes a potência são 6 dB, oito vezes são 9 dB — é isso que a escala logarítmica
faz, transformar multiplicação em soma. É o tipo de conta que você vai fazer de cabeça
depois de algumas semanas, e o Python serve aqui para conferir a intuição, não
para substituí-la.

`math` é o primeiro **módulo** que você importa. Um módulo é uma caixa de
funções prontas que não vêm carregadas por padrão; `import math` abre a caixa, e
`math.log10(...)` usa o que tem dentro.

</details>

## Escolhendo o nome das variáveis

O padrão de Python é `minusculas_com_underscore`, e o nome deve dizer a grandeza
**e a unidade** quando houver ambiguidade:

```python
p = -21.4                  # e daqui a três semanas, o que era "p"?
potencia = -21.4           # melhor
potencia_dbm = -21.4       # o que o material profissional faz
```

Isso não é preciosismo de estilo. Confundir `mW` com `dBm` é o erro que aparece
em relatório de campo, e o nome da variável é a defesa mais barata contra ele.

## Erros comuns deste capítulo

| Sintoma | Causa provável |
|---|---|
| `TypeError: can only concatenate str...` | somar texto com número sem converter |
| `NameError: name 'x' is not defined` | nome digitado errado, ou célula não executada |
| `IndentationError: unexpected indent` | espaço sobrando no começo da linha |
| resultado com dezesseis casas decimais | faltou `round(valor, 2)` |
| `ValueError: could not convert string to float` | `float("-21,4")` — em Python o separador decimal é o ponto |

## Onde isso é usado de verdade

Não há nada de escolar nestas contas: elas são o miolo de um **orçamento de enlace**
(*link budget*), o cálculo que decide se um enlace óptico ou de rádio fecha antes de
alguém subir na torre ou puxar a fibra. Potência de transmissão, ganho de antena,
perdas no percurso e sensibilidade do receptor entram todos em dB, justamente
porque na escala logarítmica o que seria multiplicação vira soma — o
[capítulo de link budget do PySDR](https://pysdr.org/content/link_budgets.html)
mostra a conta completa, e ela é feita com as mesmas quatro operações deste
capítulo.

A conversão de unidades é o outro caso. A confusão entre **MB** e **Mb** aparece em
proposta comercial, em dimensionamento de enlace e na explicação que o técnico dá
ao cliente que reclama que "contratou 100 megas e baixa 12 por segundo" — os 12 MB/s
estão certos, e são os 100 Mb/s divididos por 8.

## Resumo

- Não se declara tipo; o valor atribuído define o tipo.
- Bloco é indentação, não chave.
- Dado que vem de fora chega como texto e precisa ser convertido.
- `/` devolve float, `//` devolve inteiro, `%` dá o resto, `**` eleva.
- f-string (`f"{variavel}"`) é o jeito de montar texto.
- `import math` quando precisar de `log10`, `sqrt` e companhia.
