````python
potencia = -21.4
print(potencia, type(potencia))
potencia = "-21.4"
print(potencia, type(potencia))
````

**Preveja:** "as duas linhas imprimem a mesma coisa?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Em Python você não declara o tipo de uma variável: ele vem do valor que você

o que não existe é a declaração. `type(x)` é como você pergunta a ele qual é.

### Q2. Divisão: `/` e `//` respondem perguntas diferentes

_Leia sem rodar._

````python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(4 / 2)
````

**Preveja:** "a primeira e a última linha — sai `3` ou `3.5`? Sai `2` ou `2.0`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

As duas barras respondem perguntas diferentes. `/` é a divisão de sempre e
**sempre** devolve um número com casas decimais, mesmo quando a conta é exata
(`4 / 2` dá `2.0`). `//` descarta a parte fracionária e é o que você usa quando a
pergunta é "quantos cabem" — quantos cabos inteiros, quantas horas fechadas. E `%`
dá o que sobrou.

### Q2b. Investigue: divisão inteira com número negativo

_Leia sem rodar._

````python
print(5 // 2)
print(-5 // 2)
print(5 % 2)
print(-5 % 2)
````

**Preveja:** "`-5 // 2` dá `-2` ou `-3`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

`//` não corta a parte decimal: ele arredonda **para baixo**, em
direção ao menor. Para positivos dá no mesmo (`5 // 2` é `2`), mas `-5 // 2` dá
`-3`, e não `-2`. O resto acompanha, e por isso `-5 % 2` é `1`. Você quase nunca vai
dividir grandeza negativa em telecom, mas quando acontecer — um deslocamento de
índice, um fuso — é bom não se surpreender.

### Q3. O dado que vem de fora é texto

_Leia sem rodar._

````python
lido = "12"
print(lido * 3)
print(int(lido) * 3)
````

**Preveja:** "a primeira linha imprime `36`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

`lido` é um texto, não um número. Multiplicar texto por inteiro **repete o texto**
— por isso `"12" * 3` dá `121212`. Só depois de `int(lido)` a multiplicação vira
aritmética. Guarde isto, porque vale para o curso inteiro: tudo que chega de um
arquivo, de um equipamento ou de um formulário chega como texto, e a conversão é
sua.

### Q4. O float não é exato

_Leia sem rodar._

````python
print(0.1 + 0.2)
print(round(0.1 + 0.2, 2))
print(0.1 + 0.2 == 0.3)
````

**Preveja:** "a última linha imprime `True`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

`0.1 + 0.2` não dá exatamente `0.3` — dá um valor com uma cauda de casas decimais.
Não é defeito do Python: é como todo número com casas decimais é guardado, em
qualquer linguagem e em qualquer calculadora. A consequência prática: **não compare
resultados de conta com `==`**; arredonde antes, com `round(valor, 2)`. É por isso
que os enunciados das listas pedem o arredondamento.

### Q5. f-string: número vira relatório

_Leia sem rodar._

````python
site = "OLT-CENTRO-01"
print(f"{site}: {-21.456:.2f} dBm")
print(f"[{site:<16}]")
````

**Preveja:** "quantos caracteres tem a segunda linha impressa, contando os colchetes?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

A letra `f` antes das aspas libera o uso de `{ }` dentro do texto para encaixar
valores. Depois dos dois-pontos vem o **formato**: `:.2f` arredonda para duas casas
e `:<16` ocupa no mínimo 16 colunas alinhando à esquerda. É isso que transforma uma
pilha de `print` numa tabela que se lê de relance.
