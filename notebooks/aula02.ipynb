### Q1. A string é uma sequência

**Preveja:** "o que `[-2:]` devolve?"

_Sua previsão:_

````python
id_equip = "OLT-CENTRO-01"
print(len(id_equip))
print(id_equip[0], id_equip[-1])
print(id_equip[0:3])
print(id_equip[-2:])
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

Uma string é uma **sequência de caracteres**, e você chega a cada um pela posição.
`len` dá o tamanho, `s[0]` é o primeiro e `s[-1]` é o último — o índice negativo
conta de trás para frente, e resolve os casos em que você teria de descobrir o
tamanho antes. `s[inicio:fim]` é o **fatiamento**: devolve o pedaço que vai de
`inicio` até **antes** de `fim`.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/02-strings/#uma-string-e-uma-sequencia)

</details>

### Q1b. Investigue: as outras fatias

**Preveja:** "o que acontece quando um dos lados dos dois-pontos fica vazio?"

_Sua previsão:_

````python
print(id_equip[3:])
print(id_equip[:3])
print(id_equip[:])
print(id_equip[-3:])
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

omitir um lado significa "até a ponta": `[3:]` vai do índice 3 até o
fim, `[:3]` vai do começo até antes do 3, e `[:]` devolve a string inteira. `[-3:]`
combina as duas ideias — os três últimos caracteres, sem você precisar saber o
tamanho. Esta mesma notação vale para listas, no próximo capítulo.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/02-strings/#uma-string-e-uma-sequencia)

</details>

### Q2. A string não muda

**Preveja:** "depois do `.strip()` da segunda linha, o que tem em `bruto`?"

_Sua previsão:_

````python
bruto = "  olt-centro-01  "
bruto.strip()
print(repr(bruto))
print(repr(bruto.strip().upper()))
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

Strings são **imutáveis**: nenhum método altera o texto no lugar, todos devolvem um
texto novo. Por isso `bruto.strip()` sozinho não resolve nada — o resultado precisa
ser guardado numa variável ou encadeado, como em `bruto.strip().upper()`. Este é o
defeito mais comum das primeiras listas, e ele **não dá erro**: o programa roda e
segue com o texto sujo.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/02-strings/#os-metodos-que-resolvem-o-dia-a-dia)

</details>

### Q3. `split` devolve lista

**Preveja:** "quantos campos tem a linha de log?"

_Sua previsão:_

````python
print(id_equip.split("-"))
print(len(linha.split()))
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

`texto.split(sep)` quebra o texto nos separadores e devolve uma **lista** de
pedaços. Sem argumento, `split()` quebra em qualquer sequência de espaços, de
qualquer tamanho — que é o que se quer numa saída de comando alinhada em colunas.
Repare que a linha de log tem dez pedaços e não cinco: a mensagem final tem várias
palavras, e o `split` não sabe disso.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/02-strings/#split-e-join-o-par-que-separa-e-junta)

</details>

### Q4. O campo certo × a palavra na linha

**Preveja:** "esta linha é de um alarme crítico?"

_Sua previsão:_

````python
falso = "2026-03-02 09:00:00 INFO SWITCH-NORTE-02 alarme CRITICAL encerrado"
print("CRITICAL" in falso)
print(falso.split()[2] == "CRITICAL")
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

As duas linhas respondem perguntas diferentes. `"CRITICAL" in linha` pergunta *"a
palavra aparece em algum lugar?"* — e diz `True` para uma linha `INFO` que apenas
menciona um alarme encerrado. `linha.split()[2] == "CRITICAL"` pergunta *"o campo de
severidade é esse?"*, que é o que o relatório quer saber. **Compare o campo certo
em vez de procurar a palavra no texto todo.**

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/02-strings/#dissecando-uma-linha-de-log)

</details>

### Q5. A mensagem tem tamanho variável

**Preveja:** "`campos[4]` é a mensagem?"

_Sua previsão:_

````python
campos = linha.split()
print(campos[4])
print(" ".join(campos[4:]))
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

`campos[4]` pega só a primeira palavra da mensagem, porque a mensagem tem um número
variável de palavras. Sempre que um campo vai "daqui até o fim", a resposta é
fatiar a lista e colar de volta: `" ".join(campos[4:])`. O `" "` é o separador que
vai entre os pedaços — repare que ele reconstrói o espaço que o `split` tinha
comido.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/02-strings/#dissecando-uma-linha-de-log)

</details>

### Q6. Formatação alinhada

**Preveja:** "as duas linhas têm o mesmo comprimento?"

_Sua previsão:_

````python
print(f"[{'ONU-1':<16}{-21.456:>8.2f}]")
print(f"[{'OLT-CENTRO-01':<16}{-3.5:>8.2f}]")
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

Dentro das chaves, o que vem depois dos dois-pontos é o formato. `:<16` alinha à
esquerda ocupando no mínimo 16 colunas; `:>8.2f` alinha à direita em 8 colunas com
duas casas decimais. A largura fixa é o que faz as colunas baterem de uma linha
para a outra, e é a diferença entre um relatório que se lê e um que se decifra.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/02-strings/#f-strings-numero-que-vira-relatorio)

</details>

### Q6b. Investigue: e se o nome passar de 16 caracteres?

**Preveja:** "a coluna corta o nome ou estoura?"

_Sua previsão:_

````python
print(f"[{'OLT-CENTRO-01-SECUNDARIA':<16}{-21.456:>8.2f}]")
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

`:<16` é largura **mínima**, não máxima: se o texto for maior, ele
não é cortado — a coluna estoura e o alinhamento da tabela se perde naquela linha.
Quando os nomes têm tamanhos muito diferentes, ou você escolhe uma largura que
comporte o maior, ou corta você mesmo antes de formatar.
