## Uma string é uma sequência

````python
identificador = "OLT-CENTRO-01"

print(len(identificador))
print(identificador[0])
print(identificador[-1])
print(identificador[0:3])
print(identificador[4:10])
print(identificador[-2:])
print(identificador[3:])
print(identificador[:3])
print(identificador[:])
````

**Preveja:** "o que `[-2:]` devolve? E o que acontece quando um dos lados dos dois-pontos fica vazio?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Três ideias no exemplo:

- `len` dá o tamanho e `s[i]` dá o caractere na posição `i`, contando do zero —
  igual a C;
- **índice negativo conta de trás para frente**: `s[-1]` é o último caractere,
  `s[-2]` o penúltimo. Resolve metade dos casos em que você precisaria primeiro
  descobrir o tamanho para chegar ao fim;
- `s[inicio:fim]` é o **fatiamento**: devolve um pedaço que vai de `inicio` até
  **antes** de `fim`. Omitindo um dos lados, vale "até a ponta".

O fatiamento é a ferramenta que substitui o laço de cópia caractere a caractere.
Ele aparece o curso inteiro, e não só em strings: listas usam a mesma sintaxe.

Omitir um dos lados significa "até a ponta", e omitir os dois devolve a string
inteira — são as três últimas linhas do exemplo.

</details>

## Os métodos que resolvem o dia a dia

````python
bruto = "  olt-centro-01  "

print(repr(bruto.strip()))
print(bruto.strip().upper())
print("OLT-CENTRO-01".lower())
print("OLT-CENTRO-01".replace("-", "_"))
print("OLT-CENTRO-01".startswith("OLT"))
print("CENTRO" in "OLT-CENTRO-01")
print("OLT-CENTRO-01".count("-"))

# A string original nunca muda: todo metodo devolve uma nova.
print(repr(bruto))
````

**Preveja:** "depois do `.strip()` da segunda linha, o que tem em `bruto`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

| Método | Para quê |
|---|---|
| `.strip()` | tirar espaço e quebra de linha das pontas — obrigatório em dado lido de arquivo |
| `.upper()` / `.lower()` | normalizar antes de comparar |
| `.replace(a, b)` | trocar um trecho por outro |
| `.startswith(p)` / `.endswith(s)` | testar prefixo e sufixo |
| `x in texto` | perguntar se um trecho aparece |
| `.count(x)` | contar quantas vezes aparece |

Repare na última linha da saída: **a string original não mudou**. Strings em
Python são *imutáveis* — nenhum método altera o texto no lugar, todos devolvem um
texto novo. Por isso `bruto.strip()` sozinho não resolve nada: é preciso guardar
o resultado, ou encadear (`bruto.strip().upper()`).

!!! warning "A consequência de ser imutável"

</details>

````python
# String nao se altera no lugar: para mudar, cria-se outra.
identificador = "OLT-CENTRO-01"
identificador[0] = "X"
````

<summary><b>▶ O que aconteceu</b></summary>

`repr()`, que aparece no exemplo, mostra a string com as aspas e os espaços

## `split` e `join`: o par que separa e junta

````python
identificador = "OLT-CENTRO-01"
partes = identificador.split("-")
print(partes)
print(partes[1])

ip = "10.0.3.47"
octetos = ip.split(".")
octetos[3] = "x"
print(".".join(octetos))

# split() sem argumento quebra em qualquer espaco em branco, quantos houver.
linha = "2026-03-02  14:03:17   CRITICAL OLT-CENTRO-01"
print(linha.split())
````

**Preveja:** "quantos pedaços tem a linha de log, se o formato tem cinco campos?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

- `texto.split(sep)` devolve uma **lista** com os pedaços entre os separadores;
- `sep.join(lista)` faz o caminho inverso — e repare que o separador vem antes,
  porque é ele quem sabe como colar (`".".join(...)`, `"-".join(...)`);
- `split()` **sem argumento** é um caso especial: quebra em qualquer sequência de
  espaços em branco, de qualquer tamanho, e ignora os das pontas. É exatamente o
  que se quer para uma saída de comando com colunas alinhadas.

Esses dois métodos, juntos, são a base de todo processamento de log que você vai
fazer daqui em diante.

</details>

## Dissecando uma linha de log

````python
linha = "2026-03-02 14:03:17 CRITICAL OLT-CENTRO-01 perda de sinal na porta GPON0/1/3"

campos = linha.split()
data = campos[0]
hora = campos[1]
severidade = campos[2]
equipamento = campos[3]
mensagem = " ".join(campos[4:])

print(data, hora)
print(severidade)
print(equipamento)
print(mensagem)
print(severidade == "CRITICAL")
````

**Preveja:** "a última linha pergunta se o alarme é crítico. Ela acerta sempre?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Este é o formato que o curso vai usar como padrão: data, hora, severidade,
equipamento e mensagem, separados por espaço. Duas decisões valem comentário.

A mensagem é reconstruída com `" ".join(campos[4:])` porque ela tem um número
variável de palavras — pegar `campos[4]` daria só a primeira. Sempre que um campo
"pega daqui até o fim", o fatiamento com `join` é a resposta.

E a severidade é comparada com `campos[2] == "CRITICAL"`, não com
`"CRITICAL" in linha`. A diferença aparece numa linha como
`INFO ... alarme CRITICAL encerrado`: o `in` diria que é um alarme crítico, e
erraria. **Comparar o campo certo é mais seguro do que procurar a palavra no
texto todo** — vale a regra geral.

!!! danger "Quando a linha não tem os campos esperados"

</details>

````python
# A linha nao tem todos os campos que o codigo supoe.
linha = "2026-03-02 14:03:17"
print(linha.split()[2])
````

## f-strings: número que vira relatório

````python
site = "OLT-CENTRO-01"
potencia = -21.456

print(f"{site}: {potencia}")
print(f"{site}: {potencia:.2f} dBm")
print(f"{potencia:.0f}")
print(f"[{site:<16}]")
print(f"[{site:>16}]")
print(f"[{potencia:>8.2f}]")
print(f"{7:02d}:{5:02d}")
print(f"{0.9987:.2%}")

# :<16 e largura MINIMA: nome maior que isso nao e cortado, a coluna estoura.
print(f"[{'OLT-CENTRO-01-SECUNDARIA':<16}{-21.456:>8.2f}]")
````

**Preveja:** "todas as linhas impressas têm o mesmo comprimento?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Dentro das chaves, o que vem depois dos dois-pontos é o **formato**:

| Formato | Efeito |
|---|---|
| `:.2f` | duas casas decimais, com arredondamento |
| `:.0f` | sem casas decimais |
| `:<16` | alinhado à esquerda em 16 colunas |
| `:>8` | alinhado à direita em 8 colunas |
| `:>8.2f` | as duas coisas juntas |
| `:02d` | inteiro com zero à esquerda, ocupando 2 dígitos |
| `:.2%` | porcentagem, com duas casas |

A largura fixa é o que transforma uma pilha de `print` numa tabela legível. Um
relatório de campo com as potências desalinhadas é lido errado — e nenhuma
biblioteca é necessária para resolver isso.

Repare na última linha do exemplo: `:<16` é largura **mínima**, não máxima. Um nome
maior que isso não é cortado — a coluna estoura e o alinhamento se perde naquela
linha. Quando os nomes variam muito de tamanho, escolha uma largura que comporte o
maior.

</details>

## Erros comuns deste capítulo

| Sintoma | Causa provável |
|---|---|
| `TypeError: 'str' object does not support item assignment` | tentar `s[0] = "X"`: string é imutável |
| `IndexError: list index out of range` depois de um `split` | a linha não tinha todos os campos |
| `.strip()` "não funcionou" | o resultado não foi guardado em lugar nenhum |
| comparação falha sem motivo | sobrou espaço ou a caixa é diferente: normalize com `.strip().upper()` |
| `AttributeError: 'list' object has no attribute 'upper'` | você aplicou um método de string no resultado de um `split` |

## Onde isso é usado de verdade

O formato do log que você dissecou neste capítulo não é invenção didática: é a
estrutura do **syslog**, o protocolo padronizado pelo
[RFC 5424](https://datatracker.ietf.org/doc/html/rfc5424) pelo qual praticamente
todo equipamento de rede — roteador, switch, OLT, firewall — reporta o que acontece
com ele. Cada mensagem traz prioridade, timestamp, origem e texto livre, e quem
recebe precisa separar esses campos para fazer qualquer coisa com eles. É
exatamente por isso que existem ferramentas de
[parsing de syslog em Python](https://www.papertrail.com/solution/python-syslog/),
e todas começam onde você começou: quebrando a linha em campos.

O segundo uso é ainda mais próximo do dia a dia. A saída de um comando `show` num
equipamento é **texto feito para humano lerem**, alinhado em colunas. Para
automatizar qualquer coisa a partir dela é preciso convertê-la em dado estruturado,
e é o que fazem os *parsers* do
[Genie/pyATS](https://netboxlabs.com/blog/network-test-automation-netbox-pyats-genie/)
e o TextFSM: por baixo, eles são fatiamento, `split` e comparação de campo — os
mesmos deste capítulo, aplicados a centenas de formatos diferentes.

## Resumo

- Tudo que vem do mundo real chega como texto.
- String é sequência: `len`, índice, índice negativo e fatiamento.
- String é imutável: métodos devolvem texto novo, não alteram o original.
- `split` quebra em lista, `join` junta de volta; `split()` sem argumento é o que
  se usa em saída de comando.
- Compare o campo certo em vez de procurar a palavra na linha inteira.
- f-string com formato (`:>8.2f`, `:<16`) é o que faz relatório legível.
