### Q1. A lista cresce e carrega o próprio tamanho

**Preveja:** "quantas posições foram reservadas na declaração?"

````python
print(len(potencias), potencias[-1])
potencias.append(-30.2)
print(len(potencias), potencias)
````

### Q2. O acumulador nasce **fora** do laço

**Preveja:** "os dois imprimem a mesma coisa? Se não, o que o segundo imprime?"

_Sua previsão:_

````python
soma = 0
for p in potencias:
    soma = soma + p
print(round(soma, 2))

errado = 0
for p in potencias:
    errado = 0
    errado = errado + p
print(round(errado, 2))
````

### Q3. O candidato a extremo não começa em zero

**Preveja:** "qual é a maior potência da lista? O primeiro laço acha?"

_Sua previsão:_

````python
maior_zero = 0
for p in potencias:
    if p > maior_zero:
        maior_zero = p
print(maior_zero)

maior = potencias[0]
for p in potencias:
    if p > maior:
        maior = p
print(maior)
````

### Q4. Filtro devolve lista nova

**Preveja:** "depois do laço, quantos elementos tem `potencias`?"

_Sua previsão:_

````python
criticas = []
for p in potencias:
    if p < -25:
        criticas.append(p)
print(criticas)
print(potencias)
````

### Q5. `enumerate`: o valor não basta

**Preveja:** "qual site vai aparecer?"

_Sua previsão:_

````python
sites = ["OLT-A", "ONU-B", "OLT-C", "ONU-D", "RADIO-E"]
pior = 0
for i, p in enumerate(potencias):
    if p < potencias[pior]:
        pior = i
print(pior, sites[pior])
````

### Q5b · investigue: guardando o valor, em vez da posicao

````python
pior_valor = potencias[0]
for p in potencias:
    if p < pior_valor:
        pior_valor = p
print(pior_valor)
````

### Q6. `sorted` × `.sort()`

**Preveja:** "o que tem em `resultado`?"

_Sua previsão:_

````python
print(sorted(potencias)[:2])
print(potencias)
resultado = potencias.sort()
print(resultado)
print(potencias)
````
