### Q1. A lista cresce e carrega o próprio tamanho

**Preveja:** "quantas posições foram reservadas na declaração?"

_Sua previsão:_

````python
print(len(potencias), potencias[-1])
potencias.append(-30.2)
print(len(potencias), potencias)
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

A lista não tem tamanho fixo: `append` acrescenta no fim e ela cresce. E ela carrega
o próprio tamanho — `len(lista)` sempre diz quantos itens há, então nenhuma função
precisa receber a quantidade num parâmetro à parte. O índice negativo funciona
igual ao das strings: `lista[-1]` é o último item.

### Q2. O acumulador nasce **fora** do laço

**Preveja:** "os dois imprimem a mesma coisa? Se não, o que o segundo imprime?"

_Sua previsão:_

````python
soma = 0
for p in potencias:
    soma = soma + p
print(round(soma, 2))

errado = 0
for p in potencias:
    errado = 0
    errado = errado + p
print(round(errado, 2))
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

A variável que acumula precisa nascer **antes** do laço. Criada dentro, ela é
zerada a cada volta, e no fim vale só o que a última volta colocou nela — foi por
isso que o segundo laço imprimiu exatamente o último elemento. O pior desse erro é
que ele **não dá mensagem nenhuma**: o programa roda e devolve um número plausível.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/03-listas-lacos/#os-cinco-padroes)

</details>

### Q3. O candidato a extremo não começa em zero

**Preveja:** "qual é a maior potência da lista? O primeiro laço acha?"

_Sua previsão:_

````python
maior_zero = 0
for p in potencias:
    if p > maior_zero:
        maior_zero = p
print(maior_zero)

maior = potencias[0]
for p in potencias:
    if p > maior:
        maior = p
print(maior)
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

O candidato a extremo deve começar valendo o **primeiro elemento da lista**, e não
zero. Começar em zero funciona para valores positivos — notas, contagens — e falha
em silêncio numa lista de potências em dBm, que são todas negativas: nenhuma é
maior que zero, e o resultado sai `0`. O preço de começar pelo primeiro elemento é
ter de tratar a lista vazia, que não tem primeiro elemento.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/03-listas-lacos/#os-cinco-padroes)

</details>

### Q4. Filtro devolve lista nova

**Preveja:** "depois do laço, quantos elementos tem `potencias`?"

_Sua previsão:_

````python
criticas = []
for p in potencias:
    if p < -25:
        criticas.append(p)
print(criticas)
print(potencias)
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

Filtrar é montar uma **lista nova** com o que interessa — não é remover itens da
original. Repare na segunda linha da saída: `potencias` continua inteira. É isso que
se quer quando ela ainda vai ser usada para calcular a média ou o total.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/03-listas-lacos/#os-cinco-padroes)

### Q5. `enumerate`: o valor não basta

**Preveja:** "qual site vai aparecer?"

_Sua previsão:_

````python
sites = ["OLT-A", "ONU-B", "OLT-C", "ONU-D", "RADIO-E"]
pior = 0
for i, p in enumerate(potencias):
    if p < potencias[pior]:
        pior = i
print(pior, sites[pior])
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

`enumerate` entrega, a cada volta, a **posição** e o valor. Guardar a posição é o que
permite chegar a outra lista no mesmo índice — aqui, o nome do site. Saber que a
pior potência é `-30.2` é meia informação; saber que ela é do `RADIO-E` é a
informação inteira, porque é o equipamento que a equipe vai consertar.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/03-listas-lacos/#percorrendo-com-posicao)

</details>

### Q5b. Investigue: guardando o valor, em vez da posição

**Preveja:** "este laço acha o mesmo valor do anterior? E dá para chegar ao nome do site com ele?"

_Sua previsão:_

````python
pior_valor = potencias[0]
for p in potencias:
    if p < pior_valor:
        pior_valor = p
print(pior_valor)
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

o valor sai certo — mas só o valor. Com `-30.2` na mão você não tem
como descobrir de qual site ele veio, a não ser percorrendo tudo de novo para
procurá-lo. Guardar a **posição**, como no Q5, resolve os dois problemas de uma vez.

### Q6. `sorted` × `.sort()`

**Preveja:** "o que tem em `resultado`?"

_Sua previsão:_

````python
print(sorted(potencias)[:2])
print(potencias)
resultado = potencias.sort()
print(resultado)
print(potencias)
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

`sorted(lista)` devolve uma **lista nova** ordenada e deixa a original como estava.
`lista.sort()` ordena **no lugar** e devolve `None` — por isso `resultado` ficou
valendo `None`. A pergunta que decide entre as duas é simples: *eu ainda vou
precisar da ordem original?*

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade1-fundamentos/03-listas-lacos/#ordenar)
