## A lista

````python
potencias = [-21.4, -19.8, -27.0, -25.1]

print(potencias)
print(len(potencias))
print(potencias[0])
print(potencias[-1])
print(potencias[1:3])

potencias.append(-30.2)
print(potencias)

potencias[0] = -22.0
print(potencias)

# Lista comeca vazia e cresce -- o padrao mais comum de todos.
coletadas = []
coletadas.append(-21.0)
coletadas.append(-19.0)
print(coletadas, len(coletadas))
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

Quatro propriedades que vale marcar:

- a lista **não tem tamanho fixo**: `append` acrescenta no fim e ela cresce. Você
  não precisa saber, antes de começar a coleta, quantas medições virão;
- ela **carrega o próprio tamanho**: `len(lista)` sempre diz quantos itens há, e
  nenhuma função precisa receber a quantidade num parâmetro à parte;
- o fatiamento funciona igual ao das strings, e devolve uma lista nova;
- o índice negativo também: `lista[-1]` é o último item.

O padrão mais comum de todos está no fim do exemplo: **começar com a lista vazia
e ir acrescentando**. É assim que se acumula resultado dentro de um laço.

</details>

## Os cinco padrões

Quase tudo que se faz com uma lista é um destes cinco. Eles aparecem aqui um de
cada vez, na ordem em que você vai precisar deles — leia cada um olhando para a
saída logo abaixo antes de passar para o próximo.

### 1. Acumulador — juntar tudo num número só

````python
# Padrao 1 - ACUMULADOR: percorre tudo e junta num numero so.
potencias = [-21.4, -19.8, -27.0, -25.1]

soma = 0
for p in potencias:
    soma = soma + p

print(round(soma, 2))
print(round(soma / len(potencias), 2))
````

**Preveja:** "o que precisa existir antes do laço para esta soma funcionar?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

A variável `soma` nasce **antes** do laço, valendo zero, e cresce a cada volta. É
esse acumulador que dá total, média e qualquer outra conta que envolva a lista
inteira.

Declará-la dentro do laço é o erro clássico: ela seria zerada a cada volta, e no
fim valeria só o último item. Vale ver isso acontecendo:

</details>

````python
# O MESMO laco, com o acumulador criado no lugar errado.
potencias = [-21.4, -19.8, -27.0, -25.1]

certo = 0
for p in potencias:
    certo = certo + p
print(round(certo, 2))

errado = 0
for p in potencias:
    errado = 0             # zerado a cada volta
    errado = errado + p
print(round(errado, 2))
````

**Preveja:** "os dois laços imprimem a mesma coisa? Se não, o que o segundo imprime?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

O segundo laço imprimiu exatamente o último elemento da lista. E repare no pior
desse erro: **ele não dá mensagem nenhuma** — o programa roda e devolve um número
plausível.

</details>

### 2. Contador — quantos atendem à condição

````python
# Padrao 2 - CONTADOR: conta quantos itens atendem a uma condicao.
potencias = [-21.4, -19.8, -27.0, -25.1]

fora = 0
for p in potencias:
    if p < -25:
        fora = fora + 1

print(fora)
````

**Preveja:** "quantas medições estão abaixo de -25?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Mesma estrutura do acumulador, com duas diferenças: o `if` decide quem conta, e o
que se soma é sempre `1` — porque a pergunta é "quantos", não "quanto".

</details>

### 3. Filtro — uma lista nova, menor

````python
# Padrao 3 - FILTRO: monta uma lista NOVA, so com o que interessa.
potencias = [-21.4, -19.8, -27.0, -25.1]

criticas = []
for p in potencias:
    if p < -25:
        criticas.append(p)

print(criticas)
print(potencias)
````

**Preveja:** "depois do laço, quantos elementos tem `potencias`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

A lista nova nasce vazia antes do laço e cresce com `append`. Repare na segunda
linha da saída: **a lista original não mudou**. Filtrar não é remover — e é isso
que se quer quando ela ainda vai ser usada para calcular a média.

</details>

### 4. Extremo — o maior ou o menor

````python
# Padrao 4 - EXTREMO: guarda um candidato e vai trocando por um melhor.
potencias = [-21.4, -19.8, -27.0, -25.1]

pior = potencias[0]
for p in potencias:
    if p < pior:
        pior = p

print(pior)
````

**Preveja:** "por que o candidato começa valendo `potencias[0]`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Guarde um "candidato" e troque-o sempre que achar algo melhor. O candidato começa
valendo `potencias[0]`, e não zero.

Começar em zero funciona para uma lista de valores positivos — notas, idades,
contagens — e falha em silêncio numa lista de potências em dBm, que são todas
negativas: nenhuma seria menor que zero, e o resultado sairia `0`.

</details>

````python
# O candidato a extremo comecando em zero, numa lista de valores negativos.
potencias = [-21.4, -19.8, -27.0, -25.1]

maior_zero = 0
for p in potencias:
    if p > maior_zero:
        maior_zero = p
print(maior_zero)

maior = potencias[0]
for p in potencias:
    if p > maior:
        maior = p
print(maior)
````

**Preveja:** "qual é a maior potência da lista? O primeiro laço acha?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

O primeiro laço devolveu `0`, que não é nem uma das medições. Começar pelo
primeiro elemento funciona sempre. O preço é ter que tratar a lista vazia, que não
tem primeiro elemento.

</details>

### 5. Transformação — mesma quantidade, valores novos

````python
# Padrao 5 - TRANSFORMACAO: mesma quantidade de itens, valores novos.
potencias = [-21.4, -19.8, -27.0, -25.1]

com_margem = []
for p in potencias:
    com_margem.append(round(p - 3, 2))

print(com_margem)
print(len(potencias), len(com_margem))
````

**Preveja:** "a lista de saída tem o mesmo tamanho da de entrada?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

É o filtro **sem o `if`**: todo item entra, mas modificado. A lista de saída tem
sempre o mesmo tamanho da de entrada — a última linha confere isso.

</details>

### Os cinco de uma vez

| Padrão | Estrutura | Devolve | Exemplo em telecom |
|---|---|---|---|
| **acumulador** | variável fora do laço, `soma = soma + x` | um número | potência total, média |
| **contador** | variável em zero, `+ 1` dentro do `if` | um número | quantos alarmes críticos |
| **filtro** | lista nova, `if` e `append` | lista menor | só as linhas `CRITICAL` |
| **extremo** | candidato que vai sendo substituído | um item | o pior enlace do anel |
| **transformação** | lista nova, `append` sem `if` | lista do mesmo tamanho | aplicar margem a cada valor |

Diante de um problema novo, a primeira pergunta é **qual destes cinco ele pede**.
Quase sempre é um deles, ou dois em sequência.

!!! tip "A lista vazia não é caso raro"
    É o caso do dia em que a coleta falhou. Uma função que divide por
    `len(valores)` sem checar vai levantar `ZeroDivisionError` na frente do
    cliente. Quase todo exercício desta unidade cobra esse caso, porque em campo
    ele acontece.

## Percorrendo com posição

````python
sites = ["OLT-CENTRO-01", "ONU-SUL-4512", "OLT-NORTE-02"]
potencias = [-21.4, -27.0, -19.8]

# So o valor
for site in sites:
    print(site)

# Posicao e valor ao mesmo tempo
for i, site in enumerate(sites):
    print(i, site)

# Duas listas em paralelo
for site, potencia in zip(sites, potencias):
    print(f"{site:<16}{potencia:>8.2f}")

# A posicao do pior enlace, nao so o valor
pior = 0
for i, p in enumerate(potencias):
    if p < potencias[pior]:
        pior = i
print(pior, sites[pior])
````

**Preveja:** "qual site vai aparecer na última linha?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Três formas de percorrer, e o critério para escolher:

- `for site in sites` — quando você só precisa do **valor**. É o caso mais comum,
  e é o mais legível;
- `for i, site in enumerate(sites)` — quando precisa também da **posição**. Saber
  que a pior potência é `-27.0` é meia informação; saber que ela está na posição
  1, e que a posição 1 é a `ONU-SUL-4512`, é a informação inteira;
- `for site, potencia in zip(sites, potencias)` — quando há **duas listas em
  paralelo**, uma com os nomes e outra com os valores. `zip` anda nas duas ao
  mesmo tempo.

Em C você escreveria `for (int i = 0; i < n; i++)` e acessaria `v[i]`. Aqui isso
também existe (`for i in range(len(lista))`), mas só se usa quando nenhuma das
três formas acima serve.

</details>

## Ordenar

````python
potencias = [-21.4, -19.8, -27.0, -25.1]

# sorted() devolve uma lista nova; a original fica como estava.
print(sorted(potencias))
print(potencias)

print(sorted(potencias, reverse=True))
print(sorted(potencias)[:2])

# .sort() altera a lista no lugar e devolve None.
copia = [-21.4, -19.8, -27.0]
resultado = copia.sort()
print(copia)
print(resultado)
````

**Preveja:** "o que fica guardado em `resultado`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Duas ferramentas com a mesma finalidade e comportamentos opostos:

- **`sorted(lista)`** devolve uma **lista nova** ordenada e deixa a original
  intacta;
- **`lista.sort()`** ordena **no lugar** e devolve `None`.

A última linha do exemplo é a pegadinha que pega todo mundo uma vez:
`resultado = copia.sort()` guarda `None`, não a lista ordenada. Se você quer
guardar o resultado, é `sorted` que você quer.

Combinando `sorted` com fatiamento sai o relatório que mais se pede em telecom:
`sorted(potencias)[:2]` são os dois piores enlaces; `sorted(..., reverse=True)[:5]`
são os cinco equipamentos com mais alarmes.

</details>

## Erros comuns deste capítulo

| Sintoma | Causa provável |
|---|---|
| `IndexError: list index out of range` | índice além do fim; lembre que o último é `len-1` |
| `ZeroDivisionError: division by zero` | média de lista vazia sem tratamento |
| `TypeError: unsupported operand type(s) for +: 'int' and 'str'` | a lista veio de um arquivo e ainda é texto |
| a lista ordenada virou `None` | usou `lista.sort()` esperando o retorno |
| o resultado do laço saiu sempre igual ao último item | a variável de acumulação foi criada **dentro** do laço |
| `AttributeError: 'list' object has no attribute 'append'`… em algo que deveria ser lista | o nome foi reaproveitado para outra coisa antes |

## Onde isso é usado de verdade

Os cinco padrões são o que se faz com **contadores de desempenho**. Um elemento de
rede publica, a cada 15 minutos, séries de valores: bytes transmitidos, erros de
CRC, chamadas completadas, potência recebida. O trabalho de análise é sempre o
mesmo — somar o período, contar quantas amostras violaram um limiar, filtrar as do
enlace que interessa, achar o pico e converter a unidade. Uma operadora faz isso
sobre [contadores, alarmes, traces e CDRs](https://groupbwt.com/blog/data-extraction-for-telecom/)
para responder a uma única pergunta: *a rede está entregando o que foi prometido?*

O padrão do extremo com `enumerate` tem um nome no campo: **top-N**. "Os dez
enlaces com pior sinal", "os cinco equipamentos com mais alarmes no mês" — é o
relatório que decide para onde a equipe de campo vai na segunda-feira, e ele é o
padrão da busca do extremo aplicado a uma lista ordenada.

## Resumo

- Lista cresce com `append` e carrega o próprio tamanho.
- Cinco padrões: acumular, contar, filtrar, achar o extremo, transformar.
- A variável de acumulação nasce **antes** do laço.
- O candidato a extremo começa no primeiro elemento, não em zero.
- Lista vazia é caso real: trate-o.
- `enumerate` dá a posição, `zip` anda em duas listas.
- `sorted` devolve nova, `.sort()` altera no lugar e devolve `None`.
