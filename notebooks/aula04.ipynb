## Definindo

````python
def margem(potencia_rx, sensibilidade):
    """Folga do enlace, em dB."""
    return round(potencia_rx - sensibilidade, 2)


print(margem(-21.4, -27.0))
print(margem(-28.0, -27.0))

# A docstring fica acessivel depois -- e o que help() mostra.
print(margem.__doc__)
````

- `def nome(parametros):` abre a função; o corpo é o bloco indentado;
- não se declara o tipo dos parâmetros nem o do retorno;
- a primeira linha do corpo, entre aspas triplas, é a **docstring**: a frase que
  diz o que a função devolve. Ela não é comentário decorativo — é o que aparece
  em `help(funcao)` e na caixa de dica do editor, e é onde se registra a decisão

Neste curso, **toda função tem docstring**, e ela é escrita antes do corpo. Se
você não consegue escrever a frase, ainda não sabe o que a função faz.

## Valor padrão e retorno múltiplo

````python
def classifica(dbm, limite=-25.0):
    """Diz se a potencia esta acima do limite aceitavel."""
    if dbm >= limite:
        return "ok"
    return "fora"


print(classifica(-21.0))
print(classifica(-27.0))
print(classifica(-27.0, limite=-30.0))

# Exatamente na fronteira: a comparacao e >=, entao o valor igual ao limite passa.
print(classifica(-25.0))
print(classifica(-25.01))


def min_max(valores):
    """Devolve (menor, maior) da lista."""
    return min(valores), max(valores)


print(min_max([-21.4, -27.0, -19.8]))

menor, maior = min_max([-21.4, -27.0, -19.8])
print(menor)
print(maior)
````

**Preveja:** "as três primeiras chamadas de `classifica` — e o que sai de `min_max`: dois números ou um par?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Duas coisas que Python dá e C não dá:

**Valor padrão.** `def classifica(dbm, limite=-25.0)` deixa o segundo argumento
opcional. Quem chama escreve só `classifica(-21.0)` no caso normal, e
`classifica(-27.0, limite=-30.0)` quando precisa de outro limite. Nomear o
argumento na chamada (`limite=-30.0`) é opcional, mas deixa a linha legível — sem
isso, quem lê vê um `-30.0` solto e não sabe o que ele significa.

As duas últimas linhas do exemplo mostram os valores **exatamente na fronteira**: a
comparação é `dbm >= limite`, então o valor igual ao limite conta como `ok`. É
sempre na fronteira que uma classificação erra, e é por isso que os testes das
listas trazem um caso em cima dela.

**Retorno múltiplo.** `return min(valores), max(valores)` devolve os dois valores
de uma vez, numa **tupla**. Quem chama pode receber a tupla inteira ou
desempacotá-la em duas variáveis: `menor, maior = min_max(...)`. Em C isso exigiria
ponteiros ou uma struct.

Uma tupla é, para o que interessa agora, uma lista que não pode ser alterada.
Ela aparece naturalmente sempre que uma função devolve mais de uma informação.

</details>

## Escopo

````python
limite = -25.0          # variavel do programa principal


def conta_fora(valores):
    quantidade = 0      # so existe aqui dentro
    for valor in valores:
        if valor < limite:
            quantidade = quantidade + 1
    return quantidade


print(conta_fora([-21.0, -27.0, -30.0]))
print(limite)

# 'quantidade' nao existe aqui fora: a funcao devolveu o valor, nao a variavel.
resultado = conta_fora([-21.0])
print(resultado)
````

**Preveja:** "depois da chamada, `quantidade` ainda existe?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Variável criada dentro da função **só existe lá dentro**, e some quando a função
termina. É por isso que ela precisa `return`: o valor sai, a variável não.

A função consegue **ler** uma variável de fora (`limite`, no exemplo), e é
tentador usar isso. Não use. Uma função que depende de algo que não está na sua
lista de parâmetros não pode ser testada sozinha, nem reaproveitada em outro
script — e o dia em que alguém mudar aquela variável lá em cima, ela muda de
comportamento sem que ninguém tenha tocado nela. Tudo de que a função precisa
deve **entrar por parâmetro**.

</details>

## A decisão que a docstring precisa contar

````python
def com_margem(valores, margem):
    """Devolve uma lista NOVA, sem tocar na recebida."""
    novos = []
    for valor in valores:
        novos.append(valor + margem)
    return novos


def registra(historico, valor):
    """Altera a lista recebida, de proposito."""
    historico.append(valor)
    return historico


medicoes = [-21.0, -19.0]
print(com_margem(medicoes, -3))
print(medicoes)

print(registra(medicoes, -27.0))
print(medicoes)
````

**Preveja:** "depois das duas chamadas, o que aconteceu com a lista de quem chamou?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Leia a saída com atenção. `com_margem` devolveu uma lista nova e a original
continuou `[-21.0, -19.0]`. `registra` alterou a lista de quem chamou — depois da
chamada, `medicoes` tem um item a mais.

**Lista passada para uma função é a mesma lista**, não uma cópia. Isso é
diferente do número, que chega copiado. Então toda função que recebe uma lista
toma uma decisão:

- **devolver algo novo** e não tocar no que recebeu (é o padrão seguro, e o que
  se espera de um cálculo);
- **alterar o que recebeu**, de propósito (é o que se espera de um registro, de
  uma fila, de um histórico).

As duas são legítimas. O defeito é a função que altera sem avisar — porque quem
chamou continuou usando a lista antiga achando que ela não tinha mudado. É essa
a informação que a docstring precisa dar.

</details>

## Decompor: o exercício que importa

````python
def fora_da_faixa(potencia, limite):
    """Diz se a potencia esta abaixo do limite."""
    return potencia < limite


def linha(site, potencia):
    """Uma linha de relatorio, com as colunas alinhadas."""
    return f"{site:<16}{potencia:>8.2f}"


def relatorio(sites, potencias, limite=-25.0):
    """Relatorio dos sites fora da faixa, com o total ao final."""
    linhas = []
    for site, potencia in zip(sites, potencias):
        if fora_da_faixa(potencia, limite):
            linhas.append(linha(site, potencia))
    linhas.append(f"{len(linhas)} de {len(sites)} fora da faixa")
    return "\n".join(linhas)


sites = ["OLT-CENTRO-01", "ONU-SUL-4512", "OLT-NORTE-02", "ONU-LESTE-77"]
potencias = [-21.0, -27.5, -19.0, -28.1]
print(relatorio(sites, potencias))
````

**Preveja:** "quantas funções seriam necessárias se o critério mudasse?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

O mesmo relatório poderia ser um bloco só, com o `if` e a f-string dentro do
laço. Funcionaria. A diferença aparece depois:

- se o critério mudar (passar a considerar também potência alta demais), muda-se
  `fora_da_faixa`, **uma função de uma linha**, e nada mais;
- se o formato mudar (acrescentar a data), muda-se `linha`;
- e cada uma delas pode ser testada isoladamente, com dois ou três `assert`, sem
  montar o relatório inteiro.

Essa é a regra prática do curso: **uma função, uma decisão**. Quando a docstring
precisa de um "e" para descrever o que a função faz, provavelmente são duas.

</details>

## Testando com `assert`

As listas de exercício deste curso se corrigem sozinhas com `assert`:

```python
assert margem(-21.4, -27.0) == 5.6
```

`assert` não faz nada quando a condição é verdadeira, e levanta `AssertionError`
quando é falsa. É o teste mais barato que existe e é suficiente para função
pequena — que é exatamente o que este capítulo pede que você escreva.

## Erros comuns deste capítulo

| Sintoma | Causa provável |
|---|---|
| `TypeError: f() missing 1 required positional argument` | faltou um argumento na chamada |
| `TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'` | a função calculou mas esqueceu o `return` |
| `NameError` ao usar uma variável da função lá fora | escopo: ela morreu quando a função terminou |
| a lista do programa principal mudou sozinha | a função alterou a lista recebida |
| `IndentationError` logo depois do `def` | o corpo da função precisa estar indentado |
| a função devolve sempre o mesmo valor | `return` dentro do laço, encerrando na primeira volta |

## Onde isso é usado de verdade

As bibliotecas que a indústria usa para automação de rede são, por dentro,
exatamente isto: funções pequenas com nome claro. O
[Netmiko](https://github.com/ktbyers/netmiko) expõe `send_command` e
`send_config_set`; o NAPALM padroniza `get_facts` e `get_interfaces` para que o
mesmo código sirva a fabricantes diferentes; o Nornir roda essas funções em
paralelo sobre um inventário de centenas de equipamentos. O
[panorama do APNIC sobre essas ferramentas](https://blog.apnic.net/2023/02/13/automation-tools-paramiko-netmiko-napalm-ansible-nornir-or/)
mostra a divisão de trabalho entre elas — e ela só existe porque cada camada expõe
funções com uma responsabilidade cada.

A decisão que este capítulo insiste (**alterar o que recebeu ou devolver algo
novo?**) é a que separa, nessas bibliotecas, uma função de consulta de uma de
configuração. Confundir as duas num script que roda em cinquenta equipamentos não
é um detalhe de estilo: é a diferença entre coletar informação e mudar a rede sem
querer.

## Resumo

- `def`, corpo indentado, docstring obrigatória.
- Parâmetro com valor padrão deixa a chamada curta no caso comum.
- `return a, b` devolve uma tupla; `x, y = f()` desempacota.
- Variável local morre com a função — o que sai é o `return`.
- Tudo de que a função precisa entra por parâmetro.
- Lista chega por referência: decida se altera ou devolve nova, e diga isso na
  docstring.
- Uma função, uma decisão.
