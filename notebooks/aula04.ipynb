### Q4. Retorno múltiplo

**Preveja:** "o que sai na primeira linha — dois números ou um par?"

_Sua previsão:_

````python
def min_max(valores):
    """Devolve (menor, maior)."""
    return min(valores), max(valores)


print(min_max([-21.4, -27.0, -19.8]))
menor, maior = min_max([-21.4, -27.0, -19.8])
print(menor, maior)
````

### Q5. Escopo

_Sua previsão:_

````python
def conta(valores):
    quantidade = 0
    for v in valores:
        quantidade = quantidade + 1
    return quantidade


print(conta([1, 2, 3]))
````

### Q6. Número chega copiado, lista chega a própria

**Preveja:** "depois das duas chamadas, quanto vale `n`? E `lista`?"

_Sua previsão:_

````python
def dobra_numero(x):
    x = x * 2
    return x


def dobra_lista(v):
    for i in range(len(v)):
        v[i] = v[i] * 2
    return v


n = 10
print(dobra_numero(n), n)

lista = [1, 2, 3]
print(dobra_lista(lista), lista)
````
