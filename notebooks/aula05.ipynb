## Parte 1 — O capítulo, exemplo a exemplo

A lista resolve "quais são as medições". Ela não resolve a pergunta que o
relatório de verdade faz: **"quantos alarmes teve o OLT-CENTRO-01?"**. Para
responder isso com lista, você percorreria tudo procurando o nome — e faria isso
de novo para cada equipamento.

O dicionário guarda pares **chave → valor** e vai direto: `alarmes["OLT-CENTRO-01"]`.
É a estrutura mais usada em Python, e não por acaso — é o formato natural de todo
dado que tem nome.

## O básico

````python
# Um dicionario guarda pares chave -> valor.
equipamento = {
    "nome": "OLT-CENTRO-01",
    "ip": "10.0.1.10",
    "portas": 16,
    "em_servico": True,
}

print(equipamento["nome"])
print(equipamento["portas"])
print(len(equipamento))

# Atribuir numa chave que nao existe CRIA a chave.
equipamento["fabricante"] = "Huawei"
print(len(equipamento))

# Atribuir numa que existe SUBSTITUI o valor.
equipamento["portas"] = 24
print(equipamento["portas"])

# 'in' pergunta pela CHAVE, nunca pelo valor.
print("ip" in equipamento)
print("10.0.1.10" in equipamento)

del equipamento["fabricante"]
print(len(equipamento))
````

Acessar uma chave que não existe levanta erro:

</details>

<details>
<summary><b>▶ O que aconteceu</b></summary>

Isso é bom quando a ausência é um defeito — você quer saber. Mas na maior parte dos
casos a ausência é normal: o equipamento ainda não teve nenhum alarme, o campo
opcional não foi preenchido, é a primeira vez que aquele nome aparece.

````python
equipamento = {"nome": "OLT-CENTRO-01", "portas": 16}

# .get() devolve None quando a chave nao existe, em vez de quebrar.
print(equipamento.get("nome"))
print(equipamento.get("modelo"))

# Com um segundo argumento, devolve o padrao que voce escolher.
print(equipamento.get("modelo", "desconhecido"))
print(equipamento.get("portas", 0))

# O padrao e o que torna possivel somar sem checar antes.
contagem = {}
print(contagem.get("OLT-CENTRO-01", 0) + 1)
````

<details>
<summary><b>▶ O que aconteceu</b></summary>

`dicionario.get(chave)` devolve `None` em vez de quebrar, e
`dicionario.get(chave, padrao)` devolve o padrão que você escolher. A última linha
do exemplo é a semente do padrão mais importante deste capítulo.

!!! tip "Qual usar"
    `dicionario[chave]` quando a chave **tem que** existir — se não existir, é
    defeito e você quer o erro. `dicionario.get(chave, padrao)` quando a ausência é
    um caso previsto. Escolher um pelo outro é o que separa o script que quebra na
    frente do cliente do que esconde um problema real.

</details>

## Percorrendo

````python
potencias = {
    "OLT-CENTRO-01": -21.4,
    "ONU-SUL-4512": -27.0,
    "OLT-NORTE-02": -19.8,
}

# Percorrer o dicionario direto da as CHAVES.
for nome in potencias:
    print(nome)

# So os valores.
print(sum(potencias.values()))
print(min(potencias.values()))

# Chave e valor ao mesmo tempo -- e a forma que se usa na pratica.
for nome, dbm in potencias.items():
    print(f"{nome:<16}{dbm:>8.2f}")

print(list(potencias.keys()))
````

**Preveja:** "percorrer o dicionário direto dá as chaves, os valores, ou os dois?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

| Forma | O que o laço entrega |
|---|---|
| `for chave in dicionario` | as chaves |
| `for valor in dicionario.values()` | os valores |
| `for chave, valor in dicionario.items()` | os dois, ao mesmo tempo |

`.items()` é a que se usa na prática, porque quase todo relatório precisa do nome
**e** do número. É o mesmo desempacotamento do `enumerate` e do `zip` da Unidade 1.

A ordem do percurso é a ordem de **inserção** — a primeira chave criada é a
primeira a sair. Não confie nisso para ordenar nada: para ordem própria, use
`sorted`, como na última seção deste capítulo.

</details>

## Contar quantas vezes cada chave aparece

Esta é a operação mais útil de todo o capítulo, e ela começa com um problema
pequeno: na **primeira** vez que um equipamento aparece, não há o que somar — a
chave ainda não existe. Nas vezes seguintes, há.

O jeito direto é tratar os dois casos com um `if`:

````python
# Contar sem nenhum atalho: a primeira vez que a chave aparece e diferente das
# demais, entao o if trata os dois casos.
linhas = [
    "2026-03-02 14:03:17 CRITICAL OLT-CENTRO-01 perda de sinal",
    "2026-03-02 09:00:00 INFO SWITCH-NORTE-02 porta ativada",
    "2026-03-02 10:02:55 CRITICAL OLT-CENTRO-01 temperatura alta",
]

contagem = {}
for linha in linhas:
    nome = linha.split()[3]
    if nome in contagem:
        contagem[nome] = contagem[nome] + 1
    else:
        contagem[nome] = 1

print(contagem)
````

**Preveja:** "por que precisa de `if` aqui? O que tem de diferente na primeira vez que um nome aparece?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Funciona, e é assim que se pensa o problema. Mas repare que os dois ramos do `if`
fazem quase a mesma coisa: um soma 1 ao que havia, o outro soma 1 ao nada. Se
existisse um jeito de dizer "o valor que está lá, **ou zero se não houver nada**",
o `if` sumiria.

Existe: é o `.get(chave, 0)` da seção anterior.

```python
contagem[chave] = contagem.get(chave, 0) + 1
```

</details>

````python
# A mesma contagem do exemplo anterior, com o get resolvendo o "primeira vez".
linhas = [
    "2026-03-02 14:03:17 CRITICAL OLT-CENTRO-01 perda de sinal",
    "2026-03-02 09:00:00 INFO SWITCH-NORTE-02 porta ativada",
    "2026-03-02 10:02:55 CRITICAL OLT-CENTRO-01 temperatura alta",
    "2026-03-02 11:15:00 WARNING OLT-CENTRO-01 sinal degradado",
    "2026-03-02 09:41:12 CRITICAL ONU-SUL-4512 sem resposta",
]

por_equipamento = {}
for linha in linhas:
    nome = linha.split()[3]
    por_equipamento[nome] = por_equipamento.get(nome, 0) + 1

print(por_equipamento)

por_severidade = {}
for linha in linhas:
    severidade = linha.split()[2]
    por_severidade[severidade] = por_severidade.get(severidade, 0) + 1

print(por_severidade)
print(por_severidade["CRITICAL"])
````

<summary><b>▶ O que aconteceu</b></summary>

Leia a linha devagar: *pegue o valor atual dessa chave — ou zero, se ela ainda não
existe —, some um, e guarde de volta*. As duas versões produzem exatamente o mesmo
resultado; a segunda só escreve em uma linha o que a primeira escreve em quatro.

**Escreva a versão longa sempre que estiver em dúvida.** Ela é igualmente correta,
e é dela que a curta sai.

## Agrupar: quando o valor é uma lista

Contar responde *quantos*. Agrupar responde *quais* — e para isso o valor de cada
chave deixa de ser um número e passa a ser uma **lista**.

O raciocínio é o mesmo da contagem, com a mesma primeira vez a tratar: se a chave
ainda não existe, comece com uma lista vazia; depois, acrescente.

````python
# Contar responde "quantos". Agrupar responde "quais" -- e o valor de cada
# chave passa a ser uma LISTA.
linhas = [
    "2026-03-02 14:03:17 CRITICAL OLT-CENTRO-01 perda de sinal",
    "2026-03-02 09:00:00 INFO SWITCH-NORTE-02 porta ativada",
    "2026-03-02 10:02:55 CRITICAL OLT-CENTRO-01 temperatura alta",
]

grupos = {}
for linha in linhas:
    campos = linha.split()
    severidade = campos[2]
    equipamento = campos[3]

    if severidade not in grupos:      # primeira vez: comeca com lista vazia
        grupos[severidade] = []
    grupos[severidade].append(equipamento)

print(grupos)
print(grupos["CRITICAL"])
print(len(grupos["CRITICAL"]))
````

**Preveja:** "o valor de cada chave vai ser um número ou outra coisa?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

As duas linhas que importam são estas:

```python
if severidade not in grupos:
    grupos[severidade] = []
grupos[severidade].append(equipamento)
```

O `append` funciona porque `grupos[severidade]` **é** a lista que está dentro do
dicionário — não uma cópia dela. Alterá-la altera o dicionário.

Os dois padrões juntos — contar e agrupar — resolvem a maior parte do que se pede
num relatório de operação.

</details>

## Ordenar um dicionário

Um ranking ("os equipamentos com mais alarmes") é um problema de ordenação, e ele
vem em dois degraus.

O primeiro é saber o que ordenar. `.items()` entrega os pares:

````python
por_equipamento = {
    "OLT-CENTRO-01": 3,
    "ONU-SUL-4512": 1,
    "SWITCH-NORTE-02": 5,
}

# .items() da os pares (chave, valor).
print(list(por_equipamento.items()))

# sorted sobre esses pares ordena pela CHAVE -- ordem alfabetica de nome.
print(sorted(por_equipamento.items()))
````

**Preveja:** "`sorted` sobre `.items()` ordena por quê — pelo nome ou pela quantidade?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

`sorted` ordena esses pares pelo **primeiro** elemento de cada um — a chave. Por
isso a saída sai em ordem alfabética de nome, que não é o que um ranking quer.

O segundo degrau é dizer ao `sorted` **por qual campo** ordenar. Isso se faz com o
parâmetro `key=`, que recebe uma função:

</details>

````python
por_equipamento = {
    "OLT-CENTRO-01": 3,
    "ONU-SUL-4512": 1,
    "SWITCH-NORTE-02": 5,
}


# Uma funcao comum: recebe um par e devolve o que deve ser comparado.
def quantidade_do_par(par):
    """Devolve a quantidade de um par (nome, quantidade)."""
    return par[1]


# Ela funciona sozinha, como qualquer funcao:
print(quantidade_do_par(("OLT-CENTRO-01", 3)))

# Passando o NOME dela para o sorted (sem parenteses!), o sorted a chama para
# cada par e compara o que ela devolver.
print(sorted(por_equipamento.items(), key=quantidade_do_par))
print(sorted(por_equipamento.items(), key=quantidade_do_par, reverse=True))

# O top-N: os dois com mais alarmes.
ranking = sorted(por_equipamento.items(), key=quantidade_do_par, reverse=True)
for nome, quantos in ranking[:2]:
    print(f"{nome:<18}{quantos:>3}")
````

**Preveja:** "o que a função devolve quando chamada sozinha? E por que o `key=` vai sem parênteses?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Leia com calma, porque aqui há uma ideia nova: **uma função sendo passada como
argumento para outra**.

- `quantidade_do_par` é uma função comum. A primeira linha da saída mostra ela
  sendo chamada sozinha, como qualquer outra;
- em `key=quantidade_do_par` o nome vai **sem parênteses**. Com parênteses você
  estaria chamando a função ali mesmo e passando o resultado; sem eles, você passa
  a função em si, e quem a chama é o `sorted` — uma vez para cada par;
- para cada par, o `sorted` compara o que a função devolveu, e não o par inteiro.

Com `reverse=True` a ordem se inverte, e um fatiamento dá o **top-N**: "os dois
equipamentos com mais alarmes".

</details>

### Desempate

Quando duas chaves têm a mesma quantidade, quem vem primeiro? Sem dizer nada, a
resposta é imprevisível para quem lê o relatório:

````python
# Quando duas chaves tem a mesma quantidade, quem vem primeiro?
contagem = {"ONU-B": 3, "OLT-A": 3, "SWITCH-C": 7}


def quantidade_do_par(par):
    """Devolve a quantidade de um par (nome, quantidade)."""
    return par[1]


# Com reverse=True, o empate sai em ordem imprevisivel para quem le.
print(sorted(contagem.items(), key=quantidade_do_par, reverse=True))


# Para desempatar pelo nome, a funcao devolve uma TUPLA com os dois criterios.
# O sinal de menos inverte so a quantidade, deixando o nome em ordem normal --
# e por isso nao se usa reverse=True aqui.
def por_quantidade_e_nome(par):
    """Ordena por quantidade decrescente e, no empate, por nome crescente."""
    nome, quantidade = par
    return -quantidade, nome


print(por_quantidade_e_nome(("OLT-A", 3)))
print(sorted(contagem.items(), key=por_quantidade_e_nome))
````

**Preveja:** "com `reverse=True`, quem vem primeiro entre dois empatados?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

A solução é a função de ordenação devolver **dois critérios**, numa tupla: o
`sorted` compara o primeiro e, só em caso de empate, passa para o segundo.

O detalhe do sinal de menos: como queremos quantidade **decrescente** e nome
**crescente**, não dá para usar `reverse=True` — ele inverteria os dois. Inverter
só a quantidade, trocando-a pelo seu negativo, resolve. A linha do meio da saída
mostra o que a função devolve para um par: `(-3, 'OLT-A')`.

</details>

## Onde isso é usado de verdade

O dicionário é a forma em que o dado de rede chega quando ele já está estruturado.

Duas ferramentas centrais no mercado deixam isso explícito. Os *parsers* do
[Genie/pyATS](https://netboxlabs.com/blog/network-test-automation-netbox-pyats-genie/),
da Cisco, existem para pegar a saída em texto de um comando `show` e devolver um
**dicionário** com os mesmos dados — é o passo que transforma texto feito para
humano em algo que um programa consegue comparar, somar e verificar. E o
[NetBox](https://netboxlabs.com/docs/netbox/), que virou o padrão de *source of
truth* de inventário de rede, entrega tudo por uma API REST em JSON: no seu
programa, cada equipamento chega exatamente como o dicionário deste capítulo, com
`nome`, `ip`, `tipo` e `em_servico`.

E o padrão de contagem é, literalmente, o relatório de alarmes: agrupar as
ocorrências do dia por elemento de rede e por severidade é o primeiro número que
um centro de operações olha de manhã.

## Erros comuns deste capítulo

| Sintoma | Causa provável |
|---|---|
| `KeyError: 'X'` | chave que não existe; use `.get` se a ausência for normal |
| `TypeError: cannot use 'list' as a dict key` | chave precisa ser imutável — use tupla |
| `in` devolve `False` para algo que está no dicionário | você procurou um **valor**; `in` testa chaves |
| a contagem sai toda `1` | o dicionário foi criado **dentro** do laço |
| `TypeError: 'function' object is not subscriptable` | o `key=` recebeu a função **com** parênteses |
| `AttributeError: 'dict' object has no attribute 'append'` | dicionário não tem `append`; a atribuição na chave já cria |
| a lista de um grupo tem sempre um item só | o `dicionario[chave] = []` ficou fora do `if`, e apaga a lista a cada volta |

## Resumo

- Dicionário guarda `chave: valor`; o acesso é pela chave, não por posição.
- Atribuir cria ou substitui — a mesma sintaxe para as duas coisas.
- `in` testa **chaves**.
- `[chave]` quebra quando não existe; `.get(chave, padrao)` não.
- `.items()` é a forma de percurso que se usa na prática.
- Contar: `contagem[c] = contagem.get(c, 0) + 1` — a forma curta do `if` que
  trata a primeira ocorrência.
- Agrupar: `if c not in grupos: grupos[c] = []` e depois `grupos[c].append(item)`.
- Ranking: `sorted(d.items(), key=funcao, reverse=True)[:n]`, com a função
  passada **sem parênteses**.
