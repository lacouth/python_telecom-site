### Q1. O acesso é pela chave

_Leia sem rodar._

````python
equipamento = {"nome": "OLT-CENTRO-01", "ip": "10.0.1.10", "portas": 16}
print(equipamento["nome"])
print(len(equipamento))
print("ip" in equipamento)
print("10.0.1.10" in equipamento)
````

**Preveja:** "a última linha imprime `True` ou `False`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

O dicionário guarda pares **chave → valor**, e o acesso é pela chave, não por
posição. `len` dá a quantidade de pares. E atenção ao `in`: ele pergunta pelas
**chaves**, nunca pelos valores — por isso `"10.0.1.10" in equipamento` é `False`,
mesmo com esse IP lá dentro, já que ele é valor e não chave.

### Q4. Contar: o jeito longo primeiro

_Leia sem rodar._

````python
linhas = [
    "2026-03-02 14:03:17 CRITICAL OLT-CENTRO-01 perda de sinal",
    "2026-03-02 09:00:00 INFO SWITCH-NORTE-02 porta ativada",
    "2026-03-02 10:02:55 CRITICAL OLT-CENTRO-01 temperatura alta",
]
longo = {}
for linha in linhas:
    nome = linha.split()[3]
    if nome in longo:
        longo[nome] = longo[nome] + 1
    else:
        longo[nome] = 1
print(longo)
````

**Preveja:** "por que precisa de `if` aqui? O que tem de diferente na primeira vez que um nome aparece?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Contar tem uma dificuldade só: a **primeira** vez que uma chave aparece é diferente
das demais, porque não há o que somar ainda. O `if chave in contagem` trata os dois
casos separadamente, e está correto. Repare, porém, que os dois ramos fazem quase a
mesma coisa — é isso que a próxima célula encurta.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/05-dicionarios/#contar-quantas-vezes-cada-chave-aparece)

### Q4b. A mesma contagem, com `get` no lugar do `if`

_Leia sem rodar._

````python
contagem = {}
for linha in linhas:
    nome = linha.split()[3]
    contagem[nome] = contagem.get(nome, 0) + 1
print(contagem)
````

**Preveja:** "o resultado é o mesmo da célula anterior?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

sai idêntico. Leia a linha devagar: *pegue o valor que está nessa
chave — **ou zero, se ela ainda não existe** —, some um, guarde de volta*. O `get`
com padrão é o que faz a primeira ocorrência e a décima serem tratadas pela mesma
linha, e é o que dispensa o `if`. **Na dúvida, escreva a versão longa**: ela está
igualmente certa, e é dela que esta sai.

### Q5. Contar responde "quantos", agrupar responde "quais"

_Leia sem rodar._

````python
grupos = {}
for linha in linhas:
    campos = linha.split()
    severidade = campos[2]
    if severidade not in grupos:
        grupos[severidade] = []
    grupos[severidade].append(campos[3])
print(grupos)
````

**Preveja:** "o valor de cada chave vai ser um número ou outra coisa?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Agrupar é o mesmo raciocínio da contagem, com uma lista vazia no lugar do zero: se a
chave ainda não existe, comece com `[]`; depois, acrescente. O `.append` altera o
dicionário porque `grupos[severidade]` **é** a lista que está lá dentro, e não uma
cópia dela. Contar responde *quantos*; agrupar responde *quais*.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/05-dicionarios/#agrupar-quando-o-valor-e-uma-lista)

### Q6. Ranking: a função antes de virar argumento

_Leia sem rodar._

````python
def pelo_valor(par):
    """Devolve a quantidade de um par (nome, quantidade)."""
    return par[1]

print(pelo_valor(("OLT-CENTRO-01", 2)))
print(sorted(contagem.items()))
print(sorted(contagem.items(), key=pelo_valor, reverse=True))
````

**Preveja:** "o que a função devolve quando chamada sozinha? E as duas ordenações saem iguais ou diferentes?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

`sorted` ordena os pares de `.items()` pelo **primeiro** elemento — a chave —, que
é ordem alfabética de nome, não ranking. Para ordenar por outro campo, `key=` recebe
uma **função**: o `sorted` a chama uma vez para cada par e compara o que ela
devolver. Repare que o nome da função vai **sem parênteses** — com eles, você estaria
chamando a função ali mesmo.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/05-dicionarios/#ordenar-um-dicionario)

### Q6b. Investigue: e quando duas chaves empatam?

_Leia sem rodar._

````python
empate = {"ONU-B": 3, "OLT-A": 3, "SWITCH-C": 7}
print(sorted(empate.items(), key=pelo_valor, reverse=True))

def por_quantidade_e_nome(par):
    """Quantidade decrescente e, no empate, nome crescente."""
    nome, quantidade = par
    return -quantidade, nome

print(por_quantidade_e_nome(("OLT-A", 3)))
print(sorted(empate.items(), key=por_quantidade_e_nome))
````

**Preveja:** "com `reverse=True`, quem vem primeiro entre dois empatados?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

a ordem do empate fica imprevisível para quem lê o relatório. A
solução é a função de ordenação devolver **dois critérios**, numa tupla: o `sorted`
compara o primeiro e, só em caso de empate, passa ao segundo. Como queremos
quantidade decrescente e nome crescente, não dá para usar `reverse=True` — ele
inverteria os dois. Trocar a quantidade pelo seu negativo inverte só ela, e é o que
a segunda linha da saída mostra: `(-3, 'OLT-A')`.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/05-dicionarios/#ordenar-um-dicionario)

</details>
