````python
# ⚙️ Rode esta célula antes de tudo. Ela prepara a correção automática dos
# exercícios 🎯 — não precisa ler agora (usa coisas que só veremos mais tarde).
import math


def _mostra(argumentos):
    return ", ".join(repr(a) for a in argumentos)


def _igual(veio, esperado):
    if isinstance(esperado, float) and isinstance(veio, (int, float)):
        return math.isclose(veio, esperado, abs_tol=1e-9)
    return veio == esperado


def confere(funcao, casos):
    """Chama funcao com cada caso (argumentos, esperado) e diz se acertou."""
    certos = 0
    for numero, (argumentos, esperado) in enumerate(casos, start=1):
        chamada = f"{funcao.__name__}({_mostra(argumentos)})"
        try:
            veio = funcao(*argumentos)
        except Exception as erro:
            print(f"❌ {chamada} deu erro: {type(erro).__name__}: {erro}")
            continue
        if _igual(veio, esperado):
            certos += 1
            print(f"✅ {chamada} devolveu {veio!r}")
        elif veio is None:
            print(f"❌ {chamada} devolveu None — faltou o return?")
        else:
            print(f"❌ {chamada} devolveu {veio!r}, mas devia ser {esperado!r}")
    print(f"{certos} de {len(casos)} certos")


def confere_valor(nome, valor, esperado):
    """Diz se a variável `nome` ficou com o valor esperado."""
    if _igual(valor, esperado):
        print(f"✅ {nome} = {valor!r}")
    else:
        print(f"❌ {nome} vale {valor!r}, mas devia ser {esperado!r}")
````

## 2. Quando a chave não existe

Pedir uma chave que não está lá **quebra o programa**. Às vezes é exatamente o que
você quer; às vezes a ausência é normal — o equipamento ainda não teve alarme, o
campo opcional não veio preenchido. Para esse segundo caso existe o `get`.

📖 [capítulo 5 · `get`: a chave que pode não existir](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/05-dicionarios/#get-a-chave-que-pode-nao-existir)

> 💡 **Pense assim: a portaria do prédio.**
>
> Você pergunta ao porteiro pelo morador do 302. Com `dicionario[chave]`, se não mora
> ninguém lá, o porteiro dispara o alarme (`KeyError`) e tudo para. Com
> `dicionario.get(chave, 0)`, ele só responde "ninguém, zero" e você segue. Os dois
> são úteis: o alarme para quando a ausência é um problema; a resposta calma para
> quando a ausência é normal.

## 5. Agrupar: quando o valor é uma lista

Contar responde *quantos*. **Agrupar** responde *quais* — e o valor de cada chave
deixa de ser um número e passa a ser uma **lista**. O raciocínio é o mesmo, com a
mesma primeira vez: se a chave não existe, comece com uma lista vazia; depois,
acrescente.

📖 [capítulo 5 · Agrupar: quando o valor é uma lista](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/05-dicionarios/#agrupar-quando-o-valor-e-uma-lista)

> 💡 **Pense assim: as gavetas etiquetadas.**
>
> Agrupar é separar papéis em gavetas: cada gaveta tem uma etiqueta (`"CRITICAL"`,
> `"WARNING"`), e cada papel vai para a gaveta da sua etiqueta. Quando aparece uma
> etiqueta nova, primeiro se abre uma gaveta vazia (`grupos[severidade] = []`), e só
> então o papel entra (`append`).
