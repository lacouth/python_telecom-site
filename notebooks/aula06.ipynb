## Conjuntos: quando só importa quem está lá

````python
# Um conjunto guarda itens SEM repeticao e SEM ordem.
vistos = {"OLT-CENTRO-01", "ONU-SUL-4512", "OLT-CENTRO-01"}
print(len(vistos))
print("ONU-SUL-4512" in vistos)

# Conjunto vazio: set(), porque {} e dicionario vazio.
novos = set()
print(type(novos), type({}))

novos.add("RADIO-OESTE-01")
novos.add("RADIO-OESTE-01")
print(len(novos))

# De lista para conjunto e de volta -- o jeito curto de tirar repeticoes.
nomes = ["A", "B", "A", "C", "B"]
print(len(set(nomes)))
print(sorted(set(nomes)))
````

<summary><b>▶ O que aconteceu</b></summary>

Um `set` guarda itens **sem repetição** e **sem ordem**. As três consequências
práticas:

- repetido entra uma vez só — `{"A", "B", "A"}` tem dois itens;
- `x in conjunto` responde na hora, sem percorrer tudo, mesmo com milhares de itens
  (numa lista, essa busca percorre item por item);
- **não há índice**: `conjunto[0]` é erro, porque não existe "o primeiro".

</details>

## As operações entre conjuntos

````python
ontem = {"OLT-CENTRO-01", "OLT-NORTE-02", "ONU-SUL-4512"}
hoje = {"OLT-CENTRO-01", "OLT-NORTE-02", "RADIO-OESTE-01"}

print(sorted(hoje - ontem))      # apareceram
print(sorted(ontem - hoje))      # sumiram
print(sorted(hoje & ontem))      # continuam
print(sorted(hoje | ontem))      # todos os ja vistos
print(sorted(hoje ^ ontem))      # mudaram, para um lado ou para o outro

print(hoje == ontem)
print({"OLT-CENTRO-01"} <= hoje)  # subconjunto: esta tudo dentro?
````

**Preveja:** "antes de rodar: qual equipamento apareceu e qual sumiu?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

| Operação | Nome | A pergunta que ela responde |
|---|---|---|
| `a - b` | diferença | o que está em `a` e não está em `b` |
| `a & b` | interseção | o que está nos dois |
| `a \| b` | união | tudo, sem repetir |
| `a ^ b` | diferença simétrica | o que está num só dos dois |
| `a <= b` | subconjunto | está tudo de `a` dentro de `b`? |

Leia as duas primeiras linhas da saída de novo, porque elas são a aplicação mais
importante deste capítulo:

```python
apareceram = hoje - ontem
sumiram = ontem - hoje
```

Duas linhas comparam a coleta de hoje com a de ontem e dizem o que mudou na rede.
Sem conjunto, isso seria dois laços aninhados e uma boa chance de errar um dos
lados. Vale ver a versão longa uma vez:

</details>

````python
ontem = {"OLT-CENTRO-01", "OLT-NORTE-02", "ONU-SUL-4512"}
hoje = {"OLT-CENTRO-01", "OLT-NORTE-02", "RADIO-OESTE-01"}

# A diferenca entre conjuntos, numa linha...
print(sorted(ontem - hoje))

# ...e o mesmo resultado com lista e laco.
sumiram = []
for nome in sorted(ontem):
    if nome not in hoje:
        sumiram.append(nome)
print(sumiram)
````

<summary><b>▶ O que aconteceu</b></summary>

As duas devolvem o mesmo. O que a operação de conjuntos dá, além de brevidade, é a
impossibilidade de trocar os lados por descuido: `ontem - hoje` se lê exatamente
como a pergunta.

</details>

## Tuplas: a sequência que não muda

````python
# Tupla e uma sequencia que nao muda.
medicao = ("OLT-CENTRO-01", "2026-03-02 14:00:00", -21.4)
print(medicao[0])
print(len(medicao))

nome, momento, dbm = medicao          # desempacotamento
print(nome, dbm)

# Por nao mudar, tupla pode ser CHAVE de dicionario -- lista nao pode.
leituras = {}
leituras[("OLT-CENTRO-01", "2026-03-02")] = -21.4
leituras[("OLT-CENTRO-01", "2026-03-03")] = -21.9
print(leituras[("OLT-CENTRO-01", "2026-03-03")])
print(len(leituras))
````

Você já usou tupla sem saber o nome: é o que uma função devolve quando devolve mais
de um valor (`return menor, maior`, na Unidade 1). Use tupla quando o conjunto de
campos é fixo e ordenado — uma medição é sempre *(equipamento, momento, valor)* —,
e lista quando os itens são muitos do mesmo tipo e a quantidade varia.

A imutabilidade tem um efeito concreto: **tupla pode ser chave de dicionário**,
lista não pode. É o que permite indexar por mais de um campo ao mesmo tempo, como
`leituras[("OLT-CENTRO-01", "2026-03-02")]`.

</details>

## Estruturas dentro de estruturas

````python
# O formato em que todo inventario chega: lista de dicionarios.
inventario = [
    {"nome": "OLT-CENTRO-01", "tipo": "OLT", "portas": 16, "em_servico": True},
    {"nome": "ONU-SUL-4512", "tipo": "ONU", "portas": 1, "em_servico": True},
    {"nome": "OLT-NORTE-02", "tipo": "OLT", "portas": 8, "em_servico": False},
]

print(len(inventario))

# Chegando ao nome do primeiro equipamento, um passo por vez:
primeiro = inventario[0]      # um dicionario
print(primeiro)
print(primeiro["nome"])

# As duas linhas acima, encurtadas numa so -- e a mesma coisa:
print(inventario[0]["nome"])

# Percorrer a lista da um dicionario por vez.
for equipamento in inventario:
    if equipamento["em_servico"]:
        print(equipamento["nome"], equipamento["portas"])

# Total de portas em servico: acumulador da Unidade 1, dado da Unidade 2.
total = 0
for equipamento in inventario:
    if equipamento["em_servico"]:
        total = total + equipamento["portas"]
print(total)
````

**Preveja:** "`len(inventario)` é 3 ou 12? E o total de portas inclui a OLT fora de serviço?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

**Lista de dicionários** é o formato em que inventário, cadastro e resposta de API
chegam, sem exceção. Ele combina as duas estruturas pelo que cada uma faz melhor:
a lista dá a coleção, o dicionário dá o nome de cada campo.

A parte que assusta é o acesso a um valor lá dentro. Ele não tem mistério se você
o fizer **um passo por vez**, como no exemplo:

```python
primeiro = inventario[0]      # isto é um dicionário
primeiro["nome"]              # e este é o campo dele
```

`inventario[0]["nome"]` é só essas duas linhas escritas juntas, e se lê da
esquerda para a direita: *"o item zero da lista — que é um dicionário — e dentro
dele o campo nome"*.

**Quando a expressão ficar confusa, quebre-a em duas linhas com uma variável no
meio.** O programa faz exatamente a mesma coisa, e a leitura fica possível. Isso
não é recurso de iniciante: é o que se faz em código de produção quando o acesso
tem três ou quatro níveis.

E repare no laço que soma as portas: ele é o **acumulador da Unidade 1**, sem uma
vírgula de diferença. O que mudou foi o formato do dado, não a lógica.

E não é só o acumulador: os padrões todos continuam valendo, com o `if` olhando um
campo do dicionário.

</details>

````python
inventario = [
    {"nome": "OLT-CENTRO-01", "tipo": "OLT", "portas": 16, "em_servico": True},
    {"nome": "ONU-SUL-4512", "tipo": "ONU", "portas": 1, "em_servico": True},
    {"nome": "OLT-NORTE-02", "tipo": "OLT", "portas": 8, "em_servico": False},
]

# Filtro (Unidade 1), com o if olhando um campo do dicionario.
em_servico = []
for e in inventario:
    if e["em_servico"]:
        em_servico.append(e["nome"])
print(em_servico)

# Distintos, com conjunto.
tipos = set()
for e in inventario:
    tipos.add(e["tipo"])
print(sorted(tipos))

# Contagem (Aula 05), somando uma GRANDEZA em vez de ocorrencias.
portas_por_tipo = {}
for e in inventario:
    if e["em_servico"]:
        portas_por_tipo[e["tipo"]] = portas_por_tipo.get(e["tipo"], 0) + e["portas"]
print(portas_por_tipo)
````

**Preveja:** "qual dos cinco padrões é cada um destes três laços?"

<details>
<summary><b>▶ O que aconteceu</b></summary>

Filtro, coleta de distintos e contagem — os três do capítulo 3 e do 5. Repare na
última: ela soma **uma grandeza** (portas) em vez de contar ocorrências, e a única
diferença para a contagem de alarmes é o `+ e["portas"]` no lugar do `+ 1`.

</details>

## Consultar pelo nome, sem percorrer tudo

````python
# Procurar um equipamento pelo nome percorrendo a lista inteira funciona...
inventario = [
    {"nome": "OLT-CENTRO-01", "tipo": "OLT", "portas": 16},
    {"nome": "ONU-SUL-4512", "tipo": "ONU", "portas": 1},
    {"nome": "OLT-NORTE-02", "tipo": "OLT", "portas": 8},
]

procurado = None
for equipamento in inventario:
    if equipamento["nome"] == "OLT-NORTE-02":
        procurado = equipamento
print(procurado["portas"])

# ...mas se voce vai consultar muitas vezes, vale montar UMA VEZ um dicionario
# indexado pelo nome, e depois consultar direto.
por_nome = {}
for equipamento in inventario:
    por_nome[equipamento["nome"]] = equipamento

print(por_nome["OLT-NORTE-02"]["portas"])
print(por_nome["ONU-SUL-4512"]["tipo"])
print(sorted(por_nome))
````

**Preveja:** "e se eu precisar consultar 500 nomes numa lista de 2000 equipamentos?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

A primeira metade é o padrão da busca, da Unidade 1: percorrer a lista até achar.
Funciona — e é a resposta certa quando você precisa de um equipamento só.

A segunda metade resolve o outro caso: se você vai consultar **muitos** nomes,
percorrer a lista inteira a cada consulta fica caro. Montar uma vez o dicionário
`{nome: equipamento}` e depois consultar direto é a mesma ideia do índice
remissivo de um livro — paga-se uma vez para não procurar página por página
depois.

</details>

## Onde isso é usado de verdade

A comparação de conjuntos é a operação por trás do que o mercado chama de
**detecção de *drift*** ou verificação de conformidade: o
[NetBox](https://netboxlabs.com/docs/netbox/) guarda o estado **pretendido** da
rede — quais equipamentos deveriam existir, com quais interfaces e endereços —, e
uma coleta diária traz o estado **real**. A diferença entre os dois conjuntos é o
relatório que interessa: o que apareceu sem autorização, o que sumiu, o que mudou.
Ferramentas como o [plugin NetBox + pyATS](https://netboxlabs.com/blog/network-test-automation-netbox-pyats-genie/)
existem para produzir exatamente esse *diff*.

A lista de dicionários, por sua vez, é o formato literal em que essa API responde —
e é o mesmo do `dados/inventario.json` que o curso usa a partir do próximo
capítulo. Quando você chamar a API de um sistema de gerência na Unidade 7, o que vai
chegar na sua variável é isto aqui.

## Erros comuns deste capítulo

| Sintoma | Causa provável |
|---|---|
| `TypeError: 'set' object is not subscriptable` | conjunto não tem índice; converta com `sorted()` ou `list()` |
| `{}` criou um dicionário quando você queria conjunto | conjunto vazio é `set()` |
| a ordem dos itens mudou sozinha | conjunto não tem ordem; use `sorted` para imprimir |
| `TypeError: 'tuple' object does not support item assignment` | tupla não muda; monte outra |
| `TypeError: unhashable type: 'list'` num `set` | conjunto só aceita itens imutáveis |
| `KeyError` dentro de um laço sobre lista de dicionários | algum item não tem aquele campo; `.get` resolve |
| `TypeError: list indices must be integers` | você usou `inventario["nome"]`: o `["nome"]` é do dicionário de dentro, não da lista |

## Resumo

- `set`: sem repetição, sem ordem, busca imediata; vazio é `set()`.
- `-`, `&`, `|`, `^` comparam conjuntos — e `hoje - ontem` é o relatório de mudança.
- Tupla é sequência imutável; serve de chave composta de dicionário.
- Lista de dicionários é o formato de todo inventário e de toda resposta de API.
- Acesso aninhado se lê da esquerda para a direita, e pode sempre ser quebrado em
  duas linhas com uma variável no meio.
- Indexar a lista por nome, uma vez, evita percorrê-la a cada consulta.
