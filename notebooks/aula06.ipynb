### Q2. As duas linhas que valem a aula

_Leia sem rodar._

````python
ontem = {"OLT-CENTRO-01", "OLT-NORTE-02", "ONU-SUL-4512"}
hoje = {"OLT-CENTRO-01", "OLT-NORTE-02", "RADIO-OESTE-01"}
print(sorted(hoje - ontem))
print(sorted(ontem - hoje))
print(sorted(hoje & ontem))
````

**Preveja:** "antes de rodar: qual equipamento apareceu e qual sumiu?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Estas duas linhas são o relatório que abre o dia num centro de operações:
`hoje - ontem` diz o que **apareceu** na rede e `ontem - hoje` diz o que **sumiu**.
`&` dá o que continua nos dois. Escrito com listas e laços, isso seria meia dúzia de
linhas e uma boa chance de trocar os lados.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/06-conjuntos-tuplas/#as-operacoes-entre-conjuntos)

</details>

### Q2b. Investigue: o mesmo `ontem - hoje`, com listas e laço

_Leia sem rodar._

````python
sumiram = []
for nome in sorted(ontem):
    if nome not in hoje:
        sumiram.append(nome)
print(sumiram)
````

**Preveja:** "o resultado é o mesmo da diferença entre conjuntos?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

é o mesmo resultado, em quatro linhas em vez de uma. Escrever assim
não é errado — é o filtro da Unidade 1, e funciona. O que a operação de conjuntos
dá, além de brevidade, é a impossibilidade de trocar os lados por descuido às sete
da manhã: `ontem - hoje` se lê exatamente como a pergunta.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/06-conjuntos-tuplas/#as-operacoes-entre-conjuntos)

### Q3. Tupla: não muda, e por isso pode ser chave

_Leia sem rodar._

````python
medicao = ("OLT-CENTRO-01", "2026-03-02", -21.4)
nome, dia, dbm = medicao
print(nome, dbm)

leituras = {}
leituras[("OLT-CENTRO-01", "2026-03-02")] = -21.4
leituras[("OLT-CENTRO-01", "2026-03-03")] = -21.9
print(len(leituras), leituras[("OLT-CENTRO-01", "2026-03-03")])
````

**Preveja:** "quantas chaves tem `leituras`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Tupla é como uma lista, com duas diferenças: parênteses e **não pode ser alterada**.
Você já usava uma sem saber o nome — é o que uma função devolve quando devolve mais
de um valor. E é por não mudar que ela pode ser **chave de dicionário**, o que
permite indexar por mais de um campo ao mesmo tempo, como *(equipamento, dia)*.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/06-conjuntos-tuplas/#tuplas-a-sequencia-que-nao-muda)

### Q4. Lista de dicionários: o formato de todo inventário

_Leia sem rodar._

````python
inventario = [
    {"nome": "OLT-CENTRO-01", "tipo": "OLT", "portas": 16, "em_servico": True},
    {"nome": "ONU-SUL-4512", "tipo": "ONU", "portas": 1, "em_servico": True},
    {"nome": "OLT-NORTE-02", "tipo": "OLT", "portas": 8, "em_servico": False},
]
print(len(inventario))
primeiro = inventario[0]
print(primeiro["nome"])
print(inventario[0]["nome"])
print(inventario[2]["em_servico"])
````

**Preveja:** "`len(inventario)` é 3 ou 12?" e "quanto dá o total?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

**Lista de dicionários** é o formato em que inventário, cadastro e resposta de API
chegam. O acesso a um valor lá dentro não tem mistério se for feito um passo por
vez: `inventario[0]` é um dicionário, e `["nome"]` é o campo dele. As duas linhas do
meio imprimem a mesma coisa — `inventario[0]["nome"]` é só as duas anteriores
escritas juntas.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/06-conjuntos-tuplas/#estruturas-dentro-de-estruturas)

### Q4b. O total de portas: o acumulador, com dado novo

_Leia sem rodar._

````python
total = 0
for e in inventario:
    if e["em_servico"]:
        total = total + e["portas"]
print(total)
````

**Preveja:** "o total inclui as 8 portas da `OLT-NORTE-02`?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

não inclui — ela está com `em_servico` falso, e o `if` a deixa de
fora; por isso o total é 17 e não 25. E repare no laço: é o **acumulador** da Aula
03, idêntico. A única diferença é que o valor a somar vem de um campo do dicionário
em vez de ser o próprio item.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/06-conjuntos-tuplas/#estruturas-dentro-de-estruturas)

### Q5. Da lista para o dicionário indexado

_Leia sem rodar._

````python
por_nome = {}
for e in inventario:
    por_nome[e["nome"]] = e
print(por_nome["OLT-NORTE-02"]["portas"])
print(sorted(por_nome))
````

**Preveja:** "o que essa primeira linha constrói?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

Percorrer a lista inteira a cada consulta funciona para um equipamento; para muitos,
fica caro. Montar **uma vez** o dicionário `{nome: equipamento}` e depois consultar
direto é a mesma ideia do índice remissivo de um livro: paga-se uma vez para não
procurar página por página depois.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/06-conjuntos-tuplas/#consultar-pelo-nome-sem-percorrer-tudo)

### Q6. Os padrões da Unidade 1, agora sobre dicionários

_Leia sem rodar._

````python
em_servico = []
for e in inventario:
    if e["em_servico"]:
        em_servico.append(e["nome"])
print(em_servico)
````

**Preveja:** "qual dos cinco padrões é cada um destes três laços?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

É o **filtro** da Unidade 1, sem uma vírgula de diferença — o que mudou é que cada
item da lista agora é um dicionário, então o `if` olha um campo
(`e["em_servico"]`) e o que entra na lista nova é outro campo (`e["nome"]`).

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/06-conjuntos-tuplas/#estruturas-dentro-de-estruturas)

### Q6b. Os tipos distintos: o mesmo laço, com conjunto

````python
tipos = set()
for e in inventario:
    tipos.add(e["tipo"])
print(sorted(tipos))
````

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

saem dois, `OLT` e `ONU`: o conjunto absorve a repetição sem que

### Q6c. Portas por tipo: a contagem, somando grandeza

_Leia sem rodar._

````python
portas_por_tipo = {}
for e in inventario:
    if e["em_servico"]:
        portas_por_tipo[e["tipo"]] = portas_por_tipo.get(e["tipo"], 0) + e["portas"]
print(portas_por_tipo)
````

**Preveja:** "o valor de cada chave vai ser a quantidade de equipamentos ou a de portas?"

_Sua previsão:_

→ escreva aqui e, só então, rode a célula acima.

<details>
<summary><b>▶ O que aconteceu</b></summary>

é a de **portas** — a única diferença para a contagem da Aula 05 é o
`+ e["portas"]` no lugar do `+ 1`. Vale nomear a distinção: às vezes se contam
**ocorrências** (quantos alarmes), às vezes se acumula uma **grandeza** (quantas
portas, quantos minutos). A linha é a mesma; o que muda é o que se soma.

[Mais sobre isto no capítulo](https://lacouth.github.io/python_telecom-site/unidade2-colecoes/06-conjuntos-tuplas/#estruturas-dentro-de-estruturas)
