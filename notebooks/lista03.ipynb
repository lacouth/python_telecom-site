# Lista 03 · Listas e os padrões de laço

Cinco padrões de laço resolvem quase tudo que se faz com uma coleção de dados:
**acumular**, **contar**, **filtrar**, **buscar um extremo** e **transformar**.
Esta lista treina os cinco, um de cada vez, e termina juntando todos num
relatório.

Resista à tentação de procurar a função pronta que faz cada um deles — várias
existem, e a Unidade 6 vai usá-las à vontade. Escrever o laço agora é o que faz
você reconhecer depois o que a função pronta está fazendo por baixo.

---

**Como usar este caderno:** cada exercício tem duas células. Na primeira,
escreva a sua solução no lugar do `# TODO`. A segunda tem os testes —
rode-a e ela diz se a sua função está correta. Não altere a célula de teste.

Se um teste falhar, o Python mostra um `AssertionError` apontando a linha:
é aquele caso específico que a sua função ainda não atende.
