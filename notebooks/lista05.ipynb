# Lista 05 · Dicionários: o dado que tem nome

Esta lista percorre o dicionário na ordem em que ele é usado no trabalho: primeiro
como **cadastro** (um equipamento, seus campos), depois como **contador** (quantos
alarmes por elemento de rede) e por fim como **agrupador** e **ranking**.

Os dois padrões que valem guardar aparecem a partir do exercício 05:

```python
# contar
contagem[chave] = contagem.get(chave, 0) + 1

# agrupar
if chave not in grupos:
    grupos[chave] = []
grupos[chave].append(item)
```

Os dois resolvem a mesma dificuldade — a **primeira vez** que uma chave aparece —,
um começando do zero e o outro de uma lista vazia.

As linhas de log são as mesmas da Lista 02: o que muda é o que você faz com elas
depois de separar os campos.
