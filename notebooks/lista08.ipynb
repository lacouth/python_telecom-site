# Lista 08 · JSON, datas e exceções

Três ferramentas que, juntas, transformam um script de exercício num coletor que
roda sozinho: **JSON** (como os sistemas conversam), **datas** (como quase todo
indicador de operação é calculado) e **exceções** (como o programa sobrevive ao
dado estranho que sempre chega).

O princípio que atravessa a lista inteira aparece a partir do exercício 06: diante
de um dado ruim, o programa **não morre** e **não finge que está tudo bem** — ele
registra o que descartou e segue. Por isso quase toda função daqui devolve dois
números em vez de um.

O exercício 10 é o coletor inteiro, e é o esqueleto do projeto final.

---

**Como usar este caderno:** cada exercício tem duas células. Na primeira,
escreva a sua solução no lugar do `# TODO`. A segunda tem os testes —
rode-a e ela diz se a sua função está correta. Não altere a célula de teste.

Se um teste falhar, o Python mostra um `AssertionError` apontando a linha:
é aquele caso específico que a sua função ainda não atende.
