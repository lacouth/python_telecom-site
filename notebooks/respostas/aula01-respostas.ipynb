## 🎯 Razão em dB

```python
import math


def razao_para_db(razao):
    return round(10 * math.log10(razao), 2)
```

## 🎯 mW para dBm

```python
import math


def mw_para_dbm(potencia_mw):
    return round(10 * math.log10(potencia_mw), 2)


def sinal_ok(potencia_mw):
    dbm = mw_para_dbm(potencia_mw)
    return -27 <= dbm <= -8
```
