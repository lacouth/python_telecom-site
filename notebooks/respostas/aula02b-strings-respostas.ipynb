## 🎯 Normalizar o nome

```python
def normaliza(nome):
    return nome.strip().upper()
```

## 🎯 Mascarar o IP

```python
def mascara_ip(ip):
    octetos = ip.split(".")
    octetos[-1] = "x"
    return ".".join(octetos)
```

## 🎯 É crítico mesmo?

```python
def e_critico(linha):
    return linha.split()[2] == "CRITICAL"
```

## 🎯 Uma linha do relatório

```python
def linha_relatorio(site, potencia):
    return f"{site:<16}{potencia:>8.2f}"
```

## 🎯 O resumo do alarme

```python
def resumo(linha):
    campos = linha.split()
    hora = campos[1][:5]
    severidade = campos[2]
    equipamento = campos[3]
    mensagem = " ".join(campos[4:])
    return f"{hora} | {severidade} | {equipamento} | {mensagem}"
```
