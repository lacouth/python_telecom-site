## 🎯 As últimas medições

```python
def ultimas(medicoes, n):
    return medicoes[-n:]
```

## 🎯 Quantas fora da faixa

```python
def quantas_fora(potencias, minimo, maximo):
    quantas = 0
    for p in potencias:
        if p < minimo or p > maximo:
            quantas = quantas + 1
    return quantas
```

## 🎯 De texto para número

```python
def converte(lidas):
    numeros = []
    for valor in lidas:
        numeros.append(float(valor))
    return numeros
```

## 🎯 O melhor sinal

```python
def melhor_sinal(potencias):
    melhor = potencias[0]
    for p in potencias:
        if p > melhor:
            melhor = p
    return melhor
```

## 🎯 Qual é o pior site?

```python
def pior_site(sites, potencias):
    pior = 0
    for i, p in enumerate(potencias):
        if p < potencias[pior]:
            pior = i
    return sites[pior]
```

## 🎯 A linha do relatório

```python
def relatorio(potencias, limite):
    pior = potencias[0]
    soma = 0
    abaixo = 0
    for p in potencias:
        if p < pior:
            pior = p
        soma = soma + p
        if p < limite:
            abaixo = abaixo + 1
    media = soma / len(potencias)
    return f"pior: {pior:.1f} dBm | média: {media:.1f} dBm | abaixo de {limite}: {abaixo}"
```
