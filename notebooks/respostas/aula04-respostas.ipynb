## 🎯 Perda da fibra

```python
def perda_fibra(km, atenuacao):
    """Perda de um trecho de fibra, em dB."""
    return round(km * atenuacao, 2)
```

## 🎯 Classificar a potência

```python
def classifica_potencia(dbm, minimo=-27.0, maximo=-8.0):
    """Classifica a potência recebida como baixa, ok ou alta."""
    if dbm < minimo:
        return "baixa"
    if dbm > maximo:
        return "alta"
    return "ok"
```

## 🎯 Sem leituras inválidas

```python
def sem_invalidas(medicoes):
    """Devolve uma lista NOVA só com as leituras válidas (<= 0)."""
    validas = []
    for m in medicoes:
        if m <= 0:
            validas.append(m)
    return validas
```

## 🎯 O orçamento de potência

```python
def perda_fibra(km, atenuacao=0.35):
    """Perda de um trecho de fibra, em dB."""
    return round(km * atenuacao, 2)


def potencia_recebida(tx_dbm, km, splitter_db):
    """Potência que chega à ONU, em dBm."""
    return round(tx_dbm - perda_fibra(km) - splitter_db - 1.0, 2)


def orcamento(tx_dbm, km, splitter_db, sensibilidade=-27.0):
    """Devolve (potência recebida, margem) do enlace, em dBm e dB."""
    recebida = potencia_recebida(tx_dbm, km, splitter_db)
    margem = round(recebida - sensibilidade, 2)
    return recebida, margem
```
