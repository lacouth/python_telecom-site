## 🎯 O valor está lá?

```python
def tem_valor(equipamento, valor):
    return valor in equipamento.values()
```

## 🎯 Quem está abaixo do limite?

```python
def abaixo_do_limite(potencias, limite):
    nomes = []
    for nome, dbm in potencias.items():
        if dbm < limite:
            nomes.append(nome)
    return nomes
```

## 🎯 Alarmes por hora

```python
def conta_por_hora(linhas):
    contagem = {}
    for linha in linhas:
        hora = linha.split()[1][:2]
        contagem[hora] = contagem.get(hora, 0) + 1
    return contagem
```

## 🎯 Mensagens de cada equipamento

```python
def mensagens_por_equipamento(linhas):
    grupos = {}
    for linha in linhas:
        campos = linha.split()
        equipamento = campos[3]
        mensagem = " ".join(campos[4:])
        if equipamento not in grupos:
            grupos[equipamento] = []
        grupos[equipamento].append(mensagem)
    return grupos
```

## 🎯 Os piores do dia

```python
def por_quantidade_e_nome(par):
    """Quantidade decrescente e, no empate, nome crescente."""
    nome, quantidade = par
    return -quantidade, nome


def piores(linhas, n):
    contagem = {}
    for linha in linhas:
        nome = linha.split()[3]
        contagem[nome] = contagem.get(nome, 0) + 1
    ranking = sorted(contagem.items(), key=por_quantidade_e_nome)
    return ranking[:n]
```
