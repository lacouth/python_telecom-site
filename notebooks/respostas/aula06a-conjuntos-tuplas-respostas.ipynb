## 🎯 Quantos equipamentos distintos

```python
def quantos_distintos(nomes):
    return len(set(nomes))
```

## 🎯 O que apareceu

```python
def apareceram(ontem, hoje):
    return sorted(set(hoje) - set(ontem))
```
