## 🎯 Leitura do dia

```python
def leitura_do_dia(leituras, equipamento, dia):
    return leituras.get((equipamento, dia), "sem leitura")
```

## 🎯 Portas de um tipo

```python
def total_portas(inventario, tipo):
    total = 0
    for e in inventario:
        if e["tipo"] == tipo and e["em_servico"]:
            total = total + e["portas"]
    return total
```

## 🎯 A auditoria

```python
def auditoria(nms, campo):
    no_nms = set(nms)
    no_campo = set(campo)
    fantasmas = sorted(no_nms - no_campo)
    sem_gerencia = sorted(no_campo - no_nms)
    return fantasmas, sem_gerencia
```
