## 🎯 Contar por severidade

```python
def conta_severidade(caminho, severidade):
    quantas = 0
    with open(caminho, encoding="utf-8") as arquivo:
        for linha in arquivo:
            linha = linha.strip()
            if len(linha) == 0:
                continue
            if linha.split()[2] == severidade:
                quantas = quantas + 1
    return quantas
```

## 🎯 Separar os críticos

```python
def grava_criticos(entrada, saida):
    escritas = 0
    with open(entrada, encoding="utf-8") as origem:
        with open(saida, "w", encoding="utf-8") as destino:
            for linha in origem:
                linha = linha.strip()
                if len(linha) == 0:
                    continue
                if linha.split()[2] == "CRITICAL":
                    destino.write(linha + "\n")
                    escritas = escritas + 1
    return escritas
```

## 🎯 As potências de quem está no ar

```python
import csv


def potencias_no_ar(caminho):
    potencias = []
    with open(caminho, encoding="utf-8", newline="") as arquivo:
        for registro in csv.DictReader(arquivo):
            if registro["estado"] == "UP":
                potencias.append(float(registro["potencia_dbm"]))
    return potencias
```

## 🎯 A disponibilidade

```python
import csv


def disponibilidade(caminho):
    total = {}
    no_ar = {}
    with open(caminho, encoding="utf-8", newline="") as arquivo:
        for registro in csv.DictReader(arquivo):
            nome = registro["equipamento"]
            total[nome] = total.get(nome, 0) + 1
            if registro["estado"] == "UP":
                no_ar[nome] = no_ar.get(nome, 0) + 1
    resultado = {}
    for nome, horas in total.items():
        resultado[nome] = round(100 * no_ar.get(nome, 0) / horas, 1)
    return resultado
```
