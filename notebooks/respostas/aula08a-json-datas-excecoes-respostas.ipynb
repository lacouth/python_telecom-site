## 🎯 Quem está em serviço

```python
import json


def em_servico(texto_json):
    dados = json.loads(texto_json)
    nomes = []
    for e in dados["equipamentos"]:
        if e["em_servico"]:
            nomes.append(e["nome"])
    return nomes
```
