## 🎯 A interface da mensagem

```python
def interface_da_mensagem(mensagem):
    if not mensagem.startswith("Interface "):
        return None
    _, _, resto = mensagem.partition(" ")
    nome, _, _ = resto.partition(",")
    return nome
```

## 🎯 O momento do formato B

```python
def momento_b(mes, dia, hora, ano):
    numero_mes = MESES[mes]
    return f"{ano}-{numero_mes:02d}-{int(dia):02d} {hora}"
```

## 🎯 Um quarto fabricante

```python
def detecta_formato_v2(linha):
    if linha.startswith("{"):
        return "D"
    if linha.startswith("ts="):
        return "C"
    if linha[:3] in MESES and "%" in linha:
        return "B"
    if linha[:4].isdigit() and linha[4:5] == "-":
        return "A"
    return None
```

## 🎯 Porta válida

```python
def porta_valida(texto):
    texto = texto.strip()
    if not texto.isdigit():
        return False
    numero = int(texto)
    return 1 <= numero <= 65535
```

## 🎯 Os alarmes graves de cada equipamento

```python
GRAVES = {"EMERGENCY", "ALERT", "CRITICAL", "ERROR"}


def graves_por_equipamento(linhas):
    contagem = {}
    nao_lidas = 0
    for linha in linhas:
        alarme = le_linha(linha)
        if alarme is None:
            nao_lidas = nao_lidas + 1
            continue
        if alarme["severidade"] in GRAVES:
            nome = alarme["equipamento"]
            contagem[nome] = contagem.get(nome, 0) + 1
    return contagem, nao_lidas
```
