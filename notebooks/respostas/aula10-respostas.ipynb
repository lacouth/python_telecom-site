## 🎯 Precisa de visita?

```python
from dataclasses import dataclass


@dataclass
class Enlace:
    nome: str
    potencia_rx: float
    sensibilidade: float = -27.0

    def margem(self):
        return round(self.potencia_rx - self.sensibilidade, 2)

    def precisa_visita(self):
        return self.margem() < 3
```

## 🎯 Os enlaces para visitar

```python
def para_visitar(enlaces):
    nomes = []
    for e in enlaces:
        if e.precisa_visita():
            nomes.append(e.nome)
    return nomes
```

## 🎯 De uma linha de CSV para um Enlace

```python
def de_linha(linha):
    nome, _, potencia = linha.partition(",")
    return Enlace(nome, float(potencia))
```
