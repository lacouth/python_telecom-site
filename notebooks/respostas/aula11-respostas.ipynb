## 🎯 Horas em alerta

```python
def horas_em_alerta(df, enlace, limite):
    mascara = (df["enlace"] == enlace) & (df["potencia_dbm"] < limite)
    return int(mascara.sum())
```

## 🎯 O pico de um enlace

```python
def pico_do_enlace(df, enlace):
    do_enlace = df[df["enlace"] == enlace]
    return float(do_enlace["trafego_mbps"].max())
```

## 🎯 Tráfego médio por enlace

```python
def media_trafego(df):
    return df.groupby("enlace")["trafego_mbps"].mean().round(1).to_dict()
```

## 🎯 As horas de pico

```python
def horas_de_pico(df, n):
    por_hora = df.groupby("hora")["trafego_mbps"].mean()
    return por_hora.sort_values(ascending=False).head(n).index.tolist()
```

## 🎯 Quem descumpriu o SLA

```python
def fora_do_sla(df, sla):
    df["no_ar"] = df["estado"] == "UP"
    disp = (df.groupby("enlace")["no_ar"].mean() * 100).round(3)
    return disp[disp < sla].to_dict()
```
