## 🎯 Horas em alerta

```python
def horas_em_alerta(df, enlace, limite):
    mascara = (df["enlace"] == enlace) & (df["potencia_dbm"] < limite)
    return int(mascara.sum())
```

## 🎯 O pico de um enlace

```python
def pico_do_enlace(df, enlace):
    do_enlace = df[df["enlace"] == enlace]
    return float(do_enlace["trafego_mbps"].max())
```
