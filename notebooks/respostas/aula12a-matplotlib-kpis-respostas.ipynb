## 🎯 O dia do alerta

```python
def dia_do_alerta(df, enlace, limite):
    por_dia = df[df["enlace"] == enlace].groupby("dia")["potencia_dbm"].mean()
    abaixo = por_dia[por_dia < limite]
    if len(abaixo) == 0:
        return None
    return int(abaixo.index[0])
```
