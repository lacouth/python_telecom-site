## 🎯 Quantas horas isso dá?

```python
def horas_fora(disponibilidade, horas=744):
    return round((100 - disponibilidade) / 100 * horas, 1)
```

## 🎯 Quantas medições na faixa

```python
def quantas_na_faixa(df, minimo, maximo):
    mascara = (df["potencia_dbm"] >= minimo) & (df["potencia_dbm"] < maximo)
    return int(mascara.sum())
```

## 🎯 A figura do SLA

```python
def grafico_sla(df, sla):
    disp = df.groupby("enlace")["no_ar"].mean() * 100
    plt.figure(figsize=(8, 4))
    plt.bar(disp.index, disp.values)
    plt.axhline(sla, color="red", linestyle="--", label=f"SLA ({sla}%)")
    plt.ylim(98, 100.2)
    plt.title("Disponibilidade por enlace, março de 2026 (eixo de 98% a 100%)")
    plt.ylabel("disponibilidade (%)")
    plt.legend()
    plt.show()
    return sorted(disp[disp < sla].index.tolist())
```
