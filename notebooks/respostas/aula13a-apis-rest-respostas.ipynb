## 🎯 O tipo de um equipamento

```python
import requests


def tipo_de(api, nome):
    resposta = requests.get(api + "/api/equipamentos/" + nome, timeout=5)
    if resposta.status_code != 200:
        return None
    return resposta.json()["tipo"]
```
