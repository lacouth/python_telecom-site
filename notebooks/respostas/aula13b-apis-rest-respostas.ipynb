## 🎯 Quantos de um tipo

```python
import requests


def quantos_do_tipo(api, tipo):
    resposta = requests.get(api + "/api/equipamentos", params={"tipo": tipo}, timeout=5)
    return resposta.json()["count"]
```

## 🎯 Onde está o problema?

```python
def classifica_status(codigo):
    if 200 <= codigo <= 299:
        return "sucesso"
    if 400 <= codigo <= 499:
        return "erro no pedido"
    if 500 <= codigo <= 599:
        return "erro no servidor"
    return "outro"
```

## 🎯 O painel de críticos

```python
import requests


def painel(api):
    equipamentos = requests.get(api + "/api/equipamentos", timeout=5).json()["results"]
    resultado = {}
    for e in equipamentos:
        if e["em_servico"]:
            filtro = {"equipamento": e["nome"], "severidade": "CRITICAL"}
            resposta = requests.get(api + "/api/alarmes", params=filtro, timeout=5)
            resultado[e["nome"]] = resposta.json()["count"]
    return resultado
```
