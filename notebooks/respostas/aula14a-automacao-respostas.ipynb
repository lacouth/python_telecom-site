## 🎯 Os IPs das interfaces

```python
def ips_configurados(saida):
    ips = {}
    for linha in saida.splitlines()[1:]:
        interface, ip, ok, metodo, resto = linha.split(maxsplit=4)
        if ip != "unassigned":
            ips[interface] = ip
    return ips
```
