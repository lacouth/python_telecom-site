## 🎯 Quem responde?

```python
def alcancaveis(hosts):
    respondem = []
    for host in hosts:
        try:
            with ConnectHandler(device_type="cisco_ios", host=host,
                                username="noc", password="marenet"):
                respondem.append(host)
        except NetmikoTimeoutException:
            pass
        except NetmikoAuthenticationException:
            pass
    return respondem
```

## 🎯 A auditoria de firmware

```python
def versao_de(saida):
    for linha in saida.splitlines():
        if "Version " in linha:
            _, _, depois = linha.partition("Version ")
            versao, _, _ = depois.partition(",")
            return versao
    return None


def auditoria_firmware(hosts, versao_alvo):
    desatualizados = []
    sem_resposta = []
    for host in hosts:
        try:
            with ConnectHandler(device_type="cisco_ios", host=host,
                                username="noc", password="marenet") as conexao:
                versao = versao_de(conexao.send_command("show version"))
            if versao != versao_alvo:
                desatualizados.append(host)
        except NetmikoTimeoutException:
            sem_resposta.append(host)
        except NetmikoAuthenticationException:
            sem_resposta.append(host)
    return desatualizados, sem_resposta
```
